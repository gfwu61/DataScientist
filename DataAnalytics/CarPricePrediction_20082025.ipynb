### 2.1.4 Erstellung von Histogrammen für numerische Werte

Ein **Histogramm** stellt die **Häufigkeitsverteilung** der Werte einer numerischen Spalte grafisch dar.  
Die Werte werden in sogenannte **Bins** (Intervalle) unterteilt, und für jedes Intervall wird angezeigt, wie viele Datenpunkte darin liegen.  

Mit Histogrammen kann man auf einen Blick erkennen:
- die **Verteilung** der Daten (gleichmäßig, schief, normalverteilt, etc.),
- das Vorhandensein von **Ausreißern** oder ungewöhnlichen Werten (**Grenze: -3*std < x < +3*std**),
- **Symmetrie oder Asymmetrie** der Datenverteilung.  

Für Data Analysts ist dies sinnvoll, da Histogramme helfen, die Daten **besser zu verstehen**, die Wahl geeigneter **Transformationen** zu planen (z. B. log, sqrt) und **Vorverarbeitungsmaßnahmen** für Machine-Learning-Modelle abzuleiten.

### 2.1.10  Kategorische Merkmale vs. Preis: Boxplot

#### Der Boxplot ist ein zentrales Werkzeug in der Datenanalyse. Er visualisiert die Verteilung von Daten und macht wichtige Eigenschaften wie Lage, Streuung und mögliche Ausreißer auf einen Blick erkennbar.

#### 2.1.10.1 Interquartile Range (IQR)**

Der Interquartilbereich (IQR) beschreibt die Spannweite der mittleren 50 % der Daten.

Q1 (25. Perzentil): Wert, unter dem 25 % der Daten liegen.

Q2 (Median, 50. Perzentil): Wert, unter dem 50 % der Daten liegen.

Q3 (75. Perzentil): Wert, unter dem 75 % der Daten liegen.

Formel für IQR:

IQR=Q3−Q1

Der IQR gibt also die Spannweite des „mittleren Blocks“ der Daten an – er ignoriert die extremen Ausreißer.

Beispiel:
Wenn Q1 = 10 und Q3 = 20 → IQR = 20 − 10 = 10.

#### 2.1.10.2 Whiskers (typisch 1.5×IQR)**

Die Whiskers in einem Boxplot zeigen die Daten „außerhalb“ des mittleren Bereichs, aber ohne Ausreißer.

Standardregel:

Unteres Whisker-Limit: 
Q1−1.5×IQR

Oberes Whisker-Limit: 
Q3+1.5×IQR

Alles unterhalb von Q1 − 1.5×IQR oder oberhalb von Q3 + 1.5×IQR wird als Ausreißer markiert (oft Punkte außerhalb des Whiskers).

Beispiel:

Q1 = 10, Q3 = 20 → IQR = 10

Unteres Limit: 10 − 1.5×10 = −5

Oberes Limit: 20 + 1.5×10 = 35

Wenn deine Daten von 0 bis 40 reichen:

Whisker zeigen nur 0 bis 35

Werte >35 (z. B. 37, 40) sind Ausreißer.

#### 2.1.10.3. Visualisierung im Boxplot

          ┌─────────────┐  ← Q3 (75%)
          │             │
          │    Box      │  ← mittlere 50% (Q1–Q3)
          │             │
          └─────────────┘  ← Q1 (25%)
        Whisker:   │
                   │
                   └── Punkte = Ausreißer (>1.5×IQR)




#### Zusammenhang zwischen Whisker und 3-facher Standardabweichung
Da das Thema **Ausreißererkennung** hier behandelt wird, ist es sinnvoll einen Vergeleich der zwei Methoden zur **Ausreißererkennung** zu machen

---

#### 1. Whisker (Boxplot) – IQR-Methode

- **IQR** = Interquartilsabstand = Q3 − Q1  
- **Unteres Whisker-Limit**: `Q1 - 1.5 × IQR`  
- **Oberes Whisker-Limit**: `Q3 + 1.5 × IQR`  

**Eigenschaften:**
- Robust gegenüber Ausreißern, da nur die mittleren 50 % der Daten betrachtet werden. Das bedeutet dass Ausreißer keinen Einfluß auf IQR haben. 
- Funktioniert auch bei schiefen oder nicht-normalverteilten Daten gut.  

---

#### 2️. 3-fache Standardabweichung – Z-Score-Methode

- **Mittelwert**: μ  
- **Standardabweichung**: σ  
- **Unteres Limit**: `μ - 3σ`  
- **Oberes Limit**: `μ + 3σ`  

**Eigenschaften:**
- Annahme: Daten sind **normalverteilt**.  
- Laut 68–95–99.7-Regel liegen ~99.7 % der Werte innerhalb von ±3σ.  
- Ausreißer sind Werte >3σ vom Mittelwert entfernt.  

---

#### 3️. Vergleich / Zusammenhang

| Aspekt               | IQR (Whisker)                  | 3σ-Methode                       |
|---------------------|--------------------------------|----------------------------------|
| Basis               | Quartile (Q1, Q3)             | Mittelwert & Standardabweichung |
| Robustheit          | Sehr robust gegenüber Ausreißern | Weniger robust, Ausreißer beeinflussen σ |
| Verteilung          | Keine Annahmen                 | Annahme: Normalverteilung        |
| Typische Erkennung  | Werte >1.5×IQR vom Quartil entfernt | Werte >3σ vom Mittelwert entfernt |
| Sensitivität        | erkennen auch Moderate Ausreißer             | erkennen nur Extreme Ausreißer bei Normalverteilung |

Beispiel:





Hier sieht man den Vergleich zwischen IQR-Methode und 3σ-Methode:

IQR-Grenzen: ca. 38.8 bis 61.0

3σ-Grenzen: ca. 30.8 bis 69.7

**Interpretation:**

Die IQR-Methode ist enger → erkennt moderate Ausreißer wie den Wert 65 bereits.

Die 3σ-Methode ist viel weiter → markiert nur sehr extreme Werte wie 90 als Ausreißer, 65 und 70 bleiben drin.

**Das zeigt den Unterschied in der Sensitivität:**

IQR = empfindlicher, moderate Abweichungen gelten schon als Ausreißer.

3σ = „toleranter“, nur extreme Abweichungen zählen.

**Vergleich: Ausreißer-Erkennung mit IQR vs. 3σ**

| Wert | IQR-Methode (1.5×IQR) | 3σ-Methode |
|------|------------------------|-------------|
| 65   | ✅ Ausreißer           | ❌ kein Ausreißer |
| 70   | ✅ Ausreißer           | ✅ Ausreißer |
| 90   | ✅ Ausreißer           | ✅ Ausreißer |


**Fazit:**  
- Beide Methoden erkennen Ausreißer, aber **Whisker basiert auf der mittleren Hälfte der Daten** und ist robuster bei Schiefe oder Extremwerten.  
- Die **3σ-Regel** eignet sich besonders für normalverteilte Daten.  
- Bei symmetrischer, ungefähr normalverteilter Datenmenge sind die Grenzen oft ähnlich, aber nicht identisch.


### Python Code für Vergleich und Zusammenhang zwischen IQR und Standardabweichung:


````python
# Test Beispiel: Python Code für Vergleich / Zusammenhang zwischen IQR und Standardabweichung:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Zufällige Daten mit Normalverteilung
np.random.seed(42)
data = np.random.normal(loc=50, scale=5, size=100)

# Einige Ausreißer hinzufügen (moderate und extreme)
data = np.append(data, [65, 70, 90])  

# In DataFrame packen
df_test = pd.DataFrame(data, columns=["Werte"])

# IQR-Berechnung
Q1 = df_test["Werte"].quantile(0.25)
Q3 = df_test["Werte"].quantile(0.75)
IQR = Q3 - Q1
lower_bound_iqr = Q1 - 1.5 * IQR
upper_bound_iqr = Q3 + 1.5 * IQR

# 3σ-Berechnung
mean = df_test["Werte"].mean()
std = df_test["Werte"].std()
lower_bound_sigma = mean - 3 * std
upper_bound_sigma = mean + 3 * std

# Plot erstellen
plt.figure(figsize=(10,6))

# Boxplot für IQR
plt.subplot(2,1,1)
plt.boxplot(df_test["Werte"], vert=False)
plt.axvline(lower_bound_iqr, color="red", linestyle="--", label="IQR-Grenzen")
plt.axvline(upper_bound_iqr, color="red", linestyle="--")
plt.title("IQR-Methode (Ausreißer = Punkte außerhalb der Boxplot-Whisker)")
plt.legend()

# Histogramm mit 3σ
plt.subplot(2,1,2)
plt.hist(df_test["Werte"], bins=20, alpha=0.7, edgecolor="black")
plt.axvline(lower_bound_sigma, color="green", linestyle="--", label="3σ-Grenzen")
plt.axvline(upper_bound_sigma, color="green", linestyle="--")
plt.axvline(mean, color="blue", linestyle="-", label="Mittelwert")
plt.title("3σ-Methode")
plt.legend()

plt.tight_layout()
plt.show()

(lower_bound_iqr, upper_bound_iqr, lower_bound_sigma, upper_bound_sigma)
````

## 2.2 Explorative Datenanalyse (EDA)
Explorative Datenanalyse (EDA) ist ein methodischer Ansatz zur ersten Untersuchung von Datensätzen ohne vorgegebene Hypothesen. Ziel ist es, durch statistische und visuelle Methoden Muster, Zusammenhänge, Ausreißer und Datenstrukturen zu erkennen. Dabei hilft EDA, ein besseres Verständnis der Daten zu erlangen, Hypothesen zu generieren und die Datenqualität zu überprüfen. 

EDA verwendet häufig Streudiagramme und Korrelationsanalysen, um Zusammenhänge zwischen Variablen zu untersuchen, während die deskriptive Statistik eher einfache Zusammenfassungen und Standardvisualisierungen liefert.

### 2.2.1 Streudiagramm: Preis (y-Achse) vs. numerische Spalte (x-Achse)

Ein Streudiagramm ist für die explorative Datenanalyse (EDA) sehr nützlich, weil es:

- Zusammenhänge zwischen zwei numerischen Variablen sichtbar macht  
- Zeigt, ob eine positive, negative oder keine Korrelation vorliegt  
- Hilfreich ist, um Muster, Trends oder Ausreißer zu erkennen  
- Visuell sofort Einblicke in die Datenstruktur und mögliche Abhängigkeiten gibt  

**Kurz:**  
Ein Streudiagramm unterstützt in der EDA dabei, Beziehungen in den Daten zu verstehen und auf Basis dieser Erkenntnisse weitere Analyseschritte zu planen.

### 3.5.1 Statistische Betrachtung:
1. Was ist die Standardabweichung (σ oder std)?
 - Die Standardabweichung misst, wie stark die einzelnen Werte einer Datenmenge im Durchschnitt von deren Mittelwert (Durchschnitt) abweichen.
 
 - Kleine Standardabweichung → die Werte liegen eng um den Mittelwert herum (geringe Streuung).
 
 - Große Standardabweichung → die Werte sind weit vom Mittelwert verteilt (hohe Streuung).
 
Formel (für eine Stichprobe):



wobei
xi = einzelner Wert,
xˉ = Mittelwert,
n = Anzahl der Werte.

2. Was bedeutet „3-fache Standardabweichung“?
Wenn die Daten annähernd normalverteilt (Glockenkurve) sind, gilt die sogenannte 68–95–99,7-Regel:

 -	±1 σ → ca. 68% aller Werte liegen innerhalb einer Standardabweichung vom Mittelwert.

 -	±2 σ → ca. 95% aller Werte liegen innerhalb von zwei Standardabweichungen.

 -	±3 σ → ca. 99,7% aller Werte liegen innerhalb von drei Standardabweichungen.

3. Das heißt: die Werte außer ±3σ werden beseitigt.
Werte, die mehr als 3 Standardabweichungen vom Mittelwert entfernt sind, treten sehr selten auf (~0,3% Wahrscheinlichkeit).
In der Statistik werden diese oft als Ausreißer betrachtet.

# 4. Modeling (ML-Maschinelles Lernen) 
**Mathematische Grundlage: die Kleinste-Quadrate-Methode**
## 4.1 Der allgemeine lineare Fall
 1. model = LinearRegression()
 2. model.fit(X_train, y_train)
 
 ### 4.1.1 Ein eindimensionles Linear-Regression: 
 
 y= a*x + b; x, y: Messwerte; a: Steigerung, b: bias



 ### 4.1.2 N-dimensionles Linear-Regression: 
 
 ### die Kleinste-Quadrate-Methode(KQM)
 ### r= Xα - y; 
 m x (n+1) ; m: Anzahl der Beobachtungen(oder Anzahl der Zeilen in einer Spalte)
 
 a=(a0, a1, a2,... an), n+1-Dimension, Unbekannte;  a0: bias
 
 X: mx(n+1) dimensionale Matrix (Beobachtungen-> Xi,j: i Zeile; j: Spalte 
 
 y: m-Messwerte (Preise)
 




Ziel: die Summe durch KQM zu minimieren



Die Lösung: np.linalg.inv()




### Achtung

1. **Bedingung für eindeutige Lösung**  
   Damit die Matrix $A^T A$ stabil invertierbar ist, muss die Anzahl der **Zeilen (Beobachtungen)** stets **größer** sein als die Anzahl der **Spalten (Variablen/Features)**.  
   Ist das nicht der Fall, existiert keine eindeutige Lösung für das Gleichungssystem.

2. **Lineare Unabhängigkeit der Spalten**  
   Werden neue Spalten aus bestehenden Daten erzeugt, dürfen sie **nicht in einem proportionalen Verhältnis** zu anderen Spalten stehen. Andernfalls sind die Spalten **linear abhängig**, und die Matrix \(A^T A\) wird **singulär** (nicht invertierbar).  

   - In diesem Fall liefert `np.linalg.inv()` keine echte Inverse, sondern bestenfalls eine **Pseudo-Inverse** (z. B. über `np.linalg.pinv()`), was mathematisch problematisch ist.  
   - Beispiel:  
    - $x_1 = a \cdot x$ → ❌ **nicht erlaubt** (linear abhängig)  
    - $x_1 = \sqrt{x}$ → ✅ **erlaubt** (nicht linear abhängig)

**Fazit:**  
### Label Encoding und Ordinal Encoding

1. Vorteile von Label Encoding (**LabelEncoder** → nur alphabetisch, keine Ranglogik!)
- Kompakte Darstellung von Kategorien durch numerische Werte.  
- Einfach und effizient, besonders bei einer geringen Anzahl von Kategorien.  

2. Risiken von Label Encoding
- Kann problematisch sein, wenn das Modell die Zahlenwerte **fälschlicherweise als lineare Rangordnung** interpretiert.  
- Besonders kritisch bei vielen Kategorien, da gleiche Abstände in den Zahlen **nicht zwingend gleiche inhaltliche Unterschiede** widerspiegeln.  

3. Anwendung im aktuellen Datensatz
- In dem Datensatz gibt es meist nur **2 bis 8 Kategorien**, sodass Label Encoding hier **gut geeignet** ist und keine gravierende Verzerrung zu erwarten ist.  

4. Empfehlung: Ordinal Encoding bei Rangordnung
- Falls die Kategorien eine **fachlich begründete Reihenfolge** besitzen (z. B. *Hauptschule < Realschule < Abitur < Bachelor < Master < PhD*),  
  sollte man den **OrdinalEncoder** verwenden. 
- Damit kann die Rangordnung explizit angegeben werden, sodass die Kodierung die **inhaltliche Bedeutung korrekt widerspiegelt**.  


#### One-Hot Encoding: vermeidet dieses Problem, erzeugt aber mehr Spalten und damit größere Datensätze.
  

**In der Projektarbeit werden folgende Verfahren untersucht:**

1. Kategorische Spalten benutzt OnehotEncoder + numerische Spalten (standardisiert) → lineares Modell
2. Kategorische Spalten benutzt LabelEncoder + numerische Spalten (standardisiert) → lineares Modell
3. Kategorische Spalten benutzt LabelEncoder + numerische Spalten (standardisiert) + Ausreißer behandeln → lineares Modell

**squared_cols: 0, 1, 2; 0: keine neue Spalten;**

 - **1:**  $x_1 = \sqrt{x}$ ; falls $x < 0 \;\Rightarrow\; x_1 = -\sqrt{|x|}$  
 - **2:**  $x_2 = x^2$
   
was bedeutet "standardisiert"?

StandardScaler() ist ein Werkzeug aus scikit-learn, das numerische Features standardisiert: Es bringt alle angegebenen Spalten (Features) auf eine gemeinsame Skala, indem jeder Wert so transformiert wird:

**z = x−μσ**

**z = σx−μ**

Dabei ist **x** der Originalwert, **μ** der Mittelwert und **σ** die Standardabweichung der jeweiligen Spalte.

fit_transform() berechnet zuerst Mittelwert und Standardabweichung für jede numerische Spalte – und transformiert dann alle Werte entsprechend.

Das Ergebnis: Alle Werte der angegebenen numerischen Spalten haben danach Mittelwert=0 und Standardabweichung=1. Die Skalierung ist jetzt vergleichbar, egal ob z.B. Einkommen, Alter oder Wohnfläche vorher unterschiedlich große Zahlen waren.

### 4.3.3 Dokumentation: Scaling & Ordinal-Encoding für ML

#### 1️⃣ Numerische Werte: StandardScaler

Beim maschinellen Lernen ist es üblich, numerische Features zu standardisieren, damit sie vergleichbare Skalen haben. Dies verbessert die Stabilität und Geschwindigkeit vieler Modelle.

##### Training

```python
scaler = StandardScaler()
df_cp[numerical_columns] = scaler.fit_transform(df_cp[numerical_columns])
```

- `fit_transform()` berechnet Mittelwert (mean) und Standardabweichung (std) und wendet sie an: \$(x_{scaled} = \frac{x - mean}{std}$\)
- Die Mittelwerte und Standardabweichungen werden im `scaler` gespeichert.

##### Vorhersage / neue Daten

```python
df_test[numerical_columns] = scaler.transform(df_test[numerical_columns])
```

- Nur `transform()` wird verwendet, um die gleiche Skalierung wie beim Training zu garantieren.
- Nie `fit_transform()` auf neuen Daten verwenden!

#### 2️⃣ Kategoriale Werte: Ordinal-Encoding (LabelEncoder)

##### Training

```python
label_encoders = {}
for column in categorical_columns:
    le = LabelEncoder()
    df_cp[column] = le.fit_transform(df_cp[column])
    label_encoders[column] = le
```

- Für jede Spalte ein eigener Encoder.
- `fit_transform()` lernt die Kategorien und wandelt sie in Zahlen um.
- Die Encoder werden gespeichert, um bei Vorhersagen dieselbe Kodierung zu nutzen.

##### Vorhersage / neue Daten

```python
if label_encoders is not None:
    for column in categorical_columns:
        df_test[column] = label_encoders[column].transform(df_test[column])
```

- Die Spalten werden mit den gespeicherten Encodern transformiert.
- Neue, unbekannte Kategorien führen zu einem Fehler.

#### 3 Zusammenfassung


| Feature-Typ          | Training                              | Vorhersage                              | Warum wichtig                |
| -------------------- | ------------------------------------- | --------------------------------------- | ---------------------------- |
| Numerisch            | `fit_transform()`                     | `transform()`                           | Konsistente Standardisierung |
| Kategorial (Ordinal) | `fit_transform()` + Encoder speichern | `transform()` mit gespeichertem Encoder | Konsistente Kodierung        |



````python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd
import numpy as np

def training_carmodel(df_cp, categorical_columns, numerical_columns,
                      encoding_mode=2, test_daten_size=0.2, squared_cols=0):
    """
    Trainiert ein lineares Regressionsmodell auf Auto-Daten.

    Parameters:
    -----------
    df_cp : pd.DataFrame
        Trainingsdaten inklusive Zielspalte 'price'.
    categorical_columns : list
        Liste der kategorialen Spalten.
    numerical_columns : list
        Liste der numerischen Spalten.
    encoding_mode : int
        1 = One-Hot Encoding, 2 = Ordinal (Label) Encoding
    test_daten_size : float
        Anteil der Daten für den Testsplit.
    squared_cols : int
        0 = keine Transformation, 1 = sqrt Transformation, 2 = quadrieren

    Returns:
    --------
    model : LinearRegression
        Trainiertes Regressionsmodell
    label_encoders : dict
        Dictionary mit LabelEncodern für jede kategoriale Spalte (nur bei encoding_mode=2)
    scaler : StandardScaler
        Scaler für numerische Spalten
    mse : float
        Mean Squared Error auf Testdaten
    r2_square : float
        R^2 Score auf Testdaten
    pred_df : pd.DataFrame
        Vergleich von tatsächlichen und vorhergesagten Preisen
    """

    # ----------------------
    # Encoding
    # ----------------------
    if encoding_mode == 1:
        df_cp = pd.get_dummies(df_cp, columns=categorical_columns, drop_first=True)
        label_encoders = None
    else:
        label_encoders = {}
        for column in categorical_columns:
            le = LabelEncoder()
            df_cp[column] = le.fit_transform(df_cp[column])
            label_encoders[column] = le

    # ----------------------
    # Feature Engineering
    # ----------------------
    df_cp['power_to_weight_ratio'] = df_cp['horsepower'] / df_cp['curbweight']
    #df_cp['log_enginesize'] = np.log(df_cp['enginesize'] + 1.2)


    for column in numerical_columns:
        if squared_cols == 2:  #x^2
            df_cp[f'{column}_squared'] = df_cp[column] ** 2
        elif squared_cols == 1:  # sqrt(x)
            df_cp[f'{column}_squared'] = np.where(
                df_cp[column] <= 0,
                -np.sqrt(abs(df_cp[column])),
                np.sqrt(abs(df_cp[column]))
            )

    # ----------------------
    # Scaling
    # ----------------------
    scaler = StandardScaler()
    df_cp[numerical_columns] = scaler.fit_transform(df_cp[numerical_columns])

    # ----------------------
    # Train-Test-Split
    # ----------------------
    X = df_cp.drop(['price', 'car_ID', 'CarName'], axis=1)
    y = df_cp['price']
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_daten_size, random_state=42)

   # print("Rows:", len(df_cp), "columns", len(df_cp.columns) )
    
    # ----------------------
    # Modelltraining
    # ----------------------
    model = LinearRegression()
    model.fit(X_train, y_train)

    # ----------------------
    # Vorhersagen
    # ----------------------
    y_pred = model.predict(X_test)

    # ----------------------
    # Modellbewertung
    # ----------------------
    mse = mean_squared_error(y_test, y_pred)
    r2_square = r2_score(y_test, y_pred)

    pred_df = pd.DataFrame({
        'Actual Value': y_test,
        'Predicted Value': y_pred,
        'Difference': y_test - y_pred,
        'rel%': abs((y_test - y_pred)) / y_test * 100
    })

    #print(f"r2_score: {r2_score:.4f}, MSE: {mse:.2f}")

    return model, label_encoders, scaler, mse, r2_square, pred_df


def predict_price(df_test, model, label_encoders, scaler, categorical_columns, numerical_columns,
                  squared_cols=0):
    """
    Macht Vorhersagen für neue Auto-Daten basierend auf einem trainierten Modell.

    Parameters:
    -----------
    df_test : pd.DataFrame
        Neue Daten zum Vorhersagen
    model : LinearRegression
        Trainiertes Modell
    label_encoders : dict
        LabelEncoder pro kategorialer Spalte (nur bei Ordinal-Encoding)
    scaler : StandardScaler
        Scaler für numerische Spalten
    categorical_columns : list
        Liste der kategorialen Spalten
    numerical_columns : list
        Liste der numerischen Spalten
    squared_cols : int
        0 = keine Transformation, 1 = sqrt Transformation, 2 = quadrieren

    Returns:
    --------
    y_pred : np.ndarray
        Vorhergesagte Preise
    """

    # ----------------------
    # Encoding
    # ----------------------
    if label_encoders is not None:
        for column in categorical_columns:
            df_test[column] = label_encoders[column].transform(df_test[column])

    # ----------------------
    # Feature Engineering
    # ----------------------
    df_test['power_to_weight_ratio'] = df_test['horsepower'] / df_test['curbweight']
    #df_test['log_enginesize'] = np.log(df_test['enginesize'] + 1.2)
 

    for column in numerical_columns:
        if squared_cols == 2:
            df_test[f'{column}_squared'] = df_test[column] ** 2
        elif squared_cols == 1:
            df_test[f'{column}_squared'] = np.where(
                df_test[column] <= 0,
                -np.sqrt(abs(df_test[column])),
                np.sqrt(abs(df_test[column]))
            )

    # ----------------------
    # Scaling
    # ----------------------
    df_test[numerical_columns] = scaler.transform(df_test[numerical_columns])

    # ----------------------
    # Features für Prediction vorbereiten
    # ----------------------
    drop_cols = ['car_ID', 'CarName']
    if 'price' in df_test.columns:
        drop_cols.append('price')
    X_new = df_test.drop(drop_cols, axis=1)

    # ----------------------
    # Prediction
    # ----------------------
    y_pred = model.predict(X_new)
    return y_pred
````

### 4.3.4 Erklärung von r2_score(y_test, y_pred)

**R² (Bestimmtheitsmaß) - Erklärung**

###  1️⃣ Definition von R² (Bestimmtheitsmaß)

Für ein Test-Set gilt:

###  R² = 1 - (Σ(y_i - ŷ_i)² / Σ(y_i - ȳ)²)

#### - y_i = echte Werte  
#### - ŷ_i = vorhergesagte Werte  
#### - ȳ = Mittelwert der echten Werte

Interpretation:
- R² = 1 → perfekte Vorhersage
- R² = 0 → Modell sagt genauso gut wie Mittelwert
- R² < 0 → Modell ist schlechter als Mittelwert


###  2️⃣ Wann wird R² negativ?

### Σ(y_i - ŷ_i)² > Σ(y_i - ȳ)²

- Das passiert, wenn dein Modell die Werte so schlecht vorhersagt, dass der Mittelwert eine bessere Vorhersage liefern würde.
- Typische Ursachen:
  - Lineares Modell auf nichtlinearen Zusammenhängen
  - Zu wenige oder falsche Features
  - Ausreißer im Test-Set
  - Übermäßiges Rauschen in den Daten



Negatives R² ist ein Warnsignal, dass dein Modell nicht gut passt.
Es bedeutet nicht, dass der Wert falsch berechnet wurde.

### Antwort:

### Analyse: One-Hot-Encoding vs. Label-Encoding bei unterschiedlichen Kategorien

Du hast festgestellt, dass **One-Hot-Encoding schlechter performt** als Label-Encoding. Hier ist eine systematische Analyse.

---

### 1️⃣ Einfluss der unterschiedlichen Anzahl von Kategorien

- Beispiele:  
  - `fueltype` → 2 Kategorien (`gas`, `diesel`)  
  - `carbody` → 5 Kategorien (`convertible`, …, `wagon`)  
  - andere Spalten evtl. bis 10 Kategorien  

- **Problem bei One-Hot-Encoding:**  
  - Jede Kategorie wird in eine eigene Dummy-Spalte umgewandelt.  
  - Modelle wie **lineare Regression oder neuronale Netze** erhalten dadurch sehr unterschiedliche Dimensionen.  
  - Spalten mit vielen Dummy-Spalten können **dominieren**, während kleine Spalten unterrepräsentiert bleiben.  

- Für **Baummodelle** (DecisionTree, RandomForest) ist das weniger problematisch.

---

### 2️⃣ Einfluss von `drop_first=True`

- `drop_first=True` entfernt die **erste Kategorie jeder Spalte**, um Multikollinearität zu vermeiden.  
- Beispiel: `carbody` mit 5 Kategorien → nur 4 Dummy-Spalten bleiben.  
- **Folge:** Das Modell muss die Relation zur ausgelassenen Kategorie implizit lernen.  
- Bei **kleinen Datensätzen** oder **spärlicher Verteilung** kann das die Leistung verschlechtern.  

---

### 3️⃣ Einfluss der Anzahl der Kategorien bei Label-Encoding

- Label-Encoding wandelt jede Kategorie in eine Zahl um, z. B.:

  carbody: convertible=0, hardtop=1, hatchback=2, sedan=3, wagon=4

- Bei **kleiner Kategorieanzahl (2–10)**:  
- Unterschiede zwischen den Zahlen sind klein.  
- Lineare Modelle oder neuronale Netze können die Unterschiede leicht lernen.  
- Label-Encoding ist kompakt und oft stabiler.

- Bei vielen Kategorien (z. B. 50–100) kann Label-Encoding problematisch sein, weil das Modell eine **falsche Rangordnung** annimmt.

---

### 4️⃣ Zusammenfassung

| Faktor | Effekt auf Modellleistung |
|--------|--------------------------|
| **One-Hot, unterschiedliche Anzahl Kategorien** | Spalten mit vielen Dummies dominieren; kleine Spalten werden schwach gewichtet |
| **drop_first=True** | Kann bei kleinen Datensätzen Informationsverlust erzeugen |
| **Label-Encoding, wenige Kategorien** | Modell kann Rangordnung leicht lernen; Zahlenunterschied klein → kompakter und stabiler |

---

### 🔹 Fazit

- Dass Label-Encoding besser performt, liegt an einer Kombination von Faktoren:  
1. **Drop_first** → kleine Informationsverluste  
2. **One-Hot-Matrix** wird bei ungleichen Kategoriespalten sehr breit → schwer für Modell zu lernen  
3. **Label-Encoding mit wenigen Kategorien** → Modell kann Unterschiede gut interpretieren  

- **Empfehlung:**  
- **Lineare Modelle / kleine Datensätze:** Label-Encoding ist oft besser.  
- **Neuronale Netze oder Random Forests:** One-Hot ist sinnvoll, ggf. **drop_first=False** verwenden und Feature-Skalierung beachten.

# 5. Evaluation (Bewertung und Interpretation der Ereignisse)

## 5.1 Modellbewertung
Es gibt kategorische und numerische Daten. Für Machine-Learning müssen die Kategorie in Zhal umgewandelt werden.

Es sind hier zwei Methden One-Hot Encoding und Ordinal Encoding angewendt. 

| Methode                        | R² Score | MSE          | rel. min. | rel. max. | rel. mean. |
|--------------------------------|----------|--------------|-----------|-----------|------------|
| One-Hot-Encoding               | 0.7505   | 19,693,826.14 | 1.47      | 86.35     | 31.83      |
| Ordinal-Encoding               | 0.8660   | 10,580,608.45 | 0.22      | 72.77     | 16.96      |
| Ordinal-Encoding - Ausreißer   | 0.8771   | 12,637,217.35 | 1.77      | 57.74     | 15.15      |
| Ordinal-Encoding + Sqrt()      | 0.8919   | 11,117,848.75 | 0.61      | 41.78     | 14.71      |

Verschiedene Mothoden für Encoding werden untersucht. 
### 5.1.1 One-Hot-Encoding: 
1. One-Hot-Encoding ist gut, wenn die Kategorien wenige und gleichmäßig verteilt sind
2. Wenn die Spalten sehr unterschiedlich viele Kategorien haben, kann das Modell eine Verzerrung bekommen.
####   Die Spalten haben ungleichmäße Anzahl von Kategorien. Der Wert von R² Score=0.7505 zeigt dass One-Hot-Encoding nicht die gute Wahl ist

### 5.1.2 Ordinal-Encoding: 
1. Ordinal Encoding ist gut, wenn es eine natürliche Reihenfolge der Kategorien (Ordinalskala) gibt
   Beispiel:
#### Schulnoten: A > B > C > D > F ( 1,2,3,4,5)
#### Größe: small < medium < large
#### Zufriedenheit: low < medium < high
#### Dann macht es Sinn, sie als Zahlen darzustellen (small=0, medium=1, large=2), weil die Reihenfolge auch eine Bedeutung hat.

2. Ordinal Encoding erzeugt eine künstliche Ordnung (nach Buchstab sortiert) annehmen, die gar nicht existiert. Das beeinträchtigt die Genauigkeit des Modells
3. In Fuelsystem z.B. gibt es so viele Werte ['mpfi', '2bbl', 'mfi', '1bbl', 'spfi', '4bbl', 'idi', 'spdi'].
   Diese Werte sind nach Buchstaben sortiert und eine Zahl 1 bis 8 vergeben. Das ist nicht optimal.
3. Die Lösung wäre:
   **Die Reihenfolge für ['mpfi', '2bbl', 'mfi', '1bbl', 'spfi', '4bbl', 'idi', 'spdi'] durch Bewertung der Relevanz zu Preis ist zu bestimmen und in eine Zahlenskala umzuwandeln.**
   
   **Das ist eine Aufgabe der fachlichen Bewertung und Datenaufbereitung**
   
#### Das Ergebnis (R² Score=0.86) zeigt dass es auch ohne Reihenfolgebestimmung besseres Ergbnis als One-Hot Encoding erzielt hat

### 5.1.3 Ordinal-Encoding + Behandlung der Ausreißer: 
##### Durch statistische Analyse werden Ausreißer beseitigt. Das Ergebnis(R² Score=0.8771) verbessert sich etwas.
### 5.1.4 Ordinal-Encoding + Sqrt(): 
Das Modell wird mit **`LinearRegression()`** trainiert. 

Dieses Verfahren setzt voraus, dass die Werte in den Features **linear mit dem Zielwert (Preis)** zusammenhängen – das wäre der *Idealfall*.  

In der Praxis ist das jedoch selten gegeben. Daher ist es üblich, die numerischen Variablen je nach Datenanalyse mit **Transformationen** wie `log()`, `sqrt()` oder `x²` zu verändern, um Zusammenhänge besser darzustellen.

### 1. Textfeld: Preiseingabe
- Der Anwender gibt hier den gewünschten Preis ein.

- Der gefundene Datensatz wird in den beiden Auto-Parameter-Fenstern angezeigt, sodass der Anwender **alle Parameter** einsehen kann.  
- **Parameter-Zuordnung:**  
  - `horsepower` → Fenster „Motorleistung“  
  - `enginesize` → Fenster „Motorgröße“  

### 2. Textfeld: Motorleistung (`horsepower`)
- Motorleistung kann hier angepasst werden.  
- Dieser Wert wird für die **Preisvoraussage** als veränderliche Größe verwendet.  

### 3. Textfeld: Motorgröße (`enginesize`)
- Motorgröße kann hier angepasst werden.  
- Dieser Wert wird für die **Preisvoraussage** als veränderliche Größe verwendet.  

### 4. Textarea: Auto-Parameter
- Dient als **Informationsfenster** für den ausgewählten Datensatz.  
- Alle Parameter des Datensatzes werden hier in **zwei Bereichen** angezeigt.  

### 5. Button: Preis Berechnen
- Berechnet den Preis für die **veränderten Parameter** `horsepower` und `enginesize`.  

### 6. Textfeld: Vorausgesagter Preis
