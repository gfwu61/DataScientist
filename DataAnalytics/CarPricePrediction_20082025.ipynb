### 2.1.4 Erstellung von Histogrammen für numerische Werte

Ein **Histogramm** stellt die **Häufigkeitsverteilung** der Werte einer numerischen Spalte grafisch dar.  
Die Werte werden in sogenannte **Bins** (Intervalle) unterteilt, und für jedes Intervall wird angezeigt, wie viele Datenpunkte darin liegen.  

Mit Histogrammen kann man auf einen Blick erkennen:
- die **Verteilung** der Daten (gleichmäßig, schief, normalverteilt, etc.),
- das Vorhandensein von **Ausreißern** oder ungewöhnlichen Werten (**Grenze: -3*std < x < +3*std**),
- **Symmetrie oder Asymmetrie** der Datenverteilung.  

Für Data Analysts ist dies sinnvoll, da Histogramme helfen, die Daten **besser zu verstehen**, die Wahl geeigneter **Transformationen** zu planen (z. B. log, sqrt) und **Vorverarbeitungsmaßnahmen** für Machine-Learning-Modelle abzuleiten.

### 2.1.10  Kategorische Merkmale vs. Preis: Boxplot

#### Der Boxplot ist ein zentrales Werkzeug in der Datenanalyse. Er visualisiert die Verteilung von Daten und macht wichtige Eigenschaften wie Lage, Streuung und mögliche Ausreißer auf einen Blick erkennbar.

#### 2.1.10.1 Interquartile Range (IQR)**

Der Interquartilbereich (IQR) beschreibt die Spannweite der mittleren 50 % der Daten.

Q1 (25. Perzentil): Wert, unter dem 25 % der Daten liegen.

Q2 (Median, 50. Perzentil): Wert, unter dem 50 % der Daten liegen.

Q3 (75. Perzentil): Wert, unter dem 75 % der Daten liegen.

Formel für IQR:

IQR=Q3−Q1

Der IQR gibt also die Spannweite des „mittleren Blocks“ der Daten an – er ignoriert die extremen Ausreißer.

Beispiel:
Wenn Q1 = 10 und Q3 = 20 → IQR = 20 − 10 = 10.

#### 2.1.10.2 Whiskers (typisch 1.5×IQR)**

Die Whiskers in einem Boxplot zeigen die Daten „außerhalb“ des mittleren Bereichs, aber ohne Ausreißer.

Standardregel:

Unteres Whisker-Limit: 
Q1−1.5×IQR

Oberes Whisker-Limit: 
Q3+1.5×IQR

Alles unterhalb von Q1 − 1.5×IQR oder oberhalb von Q3 + 1.5×IQR wird als Ausreißer markiert (oft Punkte außerhalb des Whiskers).

Beispiel:

Q1 = 10, Q3 = 20 → IQR = 10

Unteres Limit: 10 − 1.5×10 = −5

Oberes Limit: 20 + 1.5×10 = 35

Wenn deine Daten von 0 bis 40 reichen:

Whisker zeigen nur 0 bis 35

Werte >35 (z. B. 37, 40) sind Ausreißer.

#### 2.1.10.3. Visualisierung im Boxplot

          ┌─────────────┐  ← Q3 (75%)
          │             │
          │    Box      │  ← mittlere 50% (Q1–Q3)
          │             │
          └─────────────┘  ← Q1 (25%)
        Whisker:   │
                   │
                   └── Punkte = Ausreißer (>1.5×IQR)




#### Zusammenhang zwischen Whisker und 3-facher Standardabweichung
Da das Thema **Ausreißererkennung** hier behandelt wird, ist es sinnvoll einen Vergeleich der zwei Methoden zur **Ausreißererkennung** zu machen

---

#### 1. Whisker (Boxplot) – IQR-Methode

- **IQR** = Interquartilsabstand = Q3 − Q1  
- **Unteres Whisker-Limit**: `Q1 - 1.5 × IQR`  
- **Oberes Whisker-Limit**: `Q3 + 1.5 × IQR`  

**Eigenschaften:**
- Robust gegenüber Ausreißern, da nur die mittleren 50 % der Daten betrachtet werden. Das bedeutet dass Ausreißer keinen Einfluß auf IQR haben. 
- Funktioniert auch bei schiefen oder nicht-normalverteilten Daten gut.  

---

#### 2️. 3-fache Standardabweichung – Z-Score-Methode

- **Mittelwert**: μ  
- **Standardabweichung**: σ  
- **Unteres Limit**: `μ - 3σ`  
- **Oberes Limit**: `μ + 3σ`  

**Eigenschaften:**
- Annahme: Daten sind **normalverteilt**.  
- Laut 68–95–99.7-Regel liegen ~99.7 % der Werte innerhalb von ±3σ.  
- Ausreißer sind Werte >3σ vom Mittelwert entfernt.  

---

#### 3️. Vergleich / Zusammenhang

| Aspekt               | IQR (Whisker)                  | 3σ-Methode                       |
|---------------------|--------------------------------|----------------------------------|
| Basis               | Quartile (Q1, Q3)             | Mittelwert & Standardabweichung |
| Robustheit          | Sehr robust gegenüber Ausreißern | Weniger robust, Ausreißer beeinflussen σ |
| Verteilung          | Keine Annahmen                 | Annahme: Normalverteilung        |
| Typische Erkennung  | Werte >1.5×IQR vom Quartil entfernt | Werte >3σ vom Mittelwert entfernt |
| Sensitivität        | erkennen auch Moderate Ausreißer             | erkennen nur Extreme Ausreißer bei Normalverteilung |

Beispiel:





Hier sieht man den Vergleich zwischen IQR-Methode und 3σ-Methode:

IQR-Grenzen: ca. 38.8 bis 61.0

3σ-Grenzen: ca. 30.8 bis 69.7

**Interpretation:**

Die IQR-Methode ist enger → erkennt moderate Ausreißer wie den Wert 65 bereits.

Die 3σ-Methode ist viel weiter → markiert nur sehr extreme Werte wie 90 als Ausreißer, 65 und 70 bleiben drin.

**Das zeigt den Unterschied in der Sensitivität:**

IQR = empfindlicher, moderate Abweichungen gelten schon als Ausreißer.

3σ = „toleranter“, nur extreme Abweichungen zählen.

**Vergleich: Ausreißer-Erkennung mit IQR vs. 3σ**

| Wert | IQR-Methode (1.5×IQR) | 3σ-Methode |
|------|------------------------|-------------|
| 65   | ✅ Ausreißer           | ❌ kein Ausreißer |
| 70   | ✅ Ausreißer           | ✅ Ausreißer |
| 90   | ✅ Ausreißer           | ✅ Ausreißer |


**Fazit:**  
- Beide Methoden erkennen Ausreißer, aber **Whisker basiert auf der mittleren Hälfte der Daten** und ist robuster bei Schiefe oder Extremwerten.  
- Die **3σ-Regel** eignet sich besonders für normalverteilte Daten.  
- Bei symmetrischer, ungefähr normalverteilter Datenmenge sind die Grenzen oft ähnlich, aber nicht identisch.


### Python Code für Vergleich und Zusammenhang zwischen IQR und Standardabweichung:


````python
# Test Beispiel: Python Code für Vergleich / Zusammenhang zwischen IQR und Standardabweichung:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Zufällige Daten mit Normalverteilung
np.random.seed(42)
data = np.random.normal(loc=50, scale=5, size=100)

# Einige Ausreißer hinzufügen (moderate und extreme)
data = np.append(data, [65, 70, 90])  

# In DataFrame packen
df_test = pd.DataFrame(data, columns=["Werte"])

# IQR-Berechnung
Q1 = df_test["Werte"].quantile(0.25)
Q3 = df_test["Werte"].quantile(0.75)
IQR = Q3 - Q1
lower_bound_iqr = Q1 - 1.5 * IQR
upper_bound_iqr = Q3 + 1.5 * IQR

print("lower_bound_iqr", lower_bound_iqr)
print("upper_bound_iqr", upper_bound_iqr)

# 3σ-Berechnung
mean = df_test["Werte"].mean()
std = df_test["Werte"].std()
lower_bound_sigma = mean - 3 * std
upper_bound_sigma = mean + 3 * std

print("lower_bound_sigma", lower_bound_sigma)
print("upper_bound_sigma", upper_bound_sigma)

# Plot erstellen
plt.figure(figsize=(10,6))

# Boxplot für IQR
plt.subplot(2,1,1)
plt.boxplot(df_test["Werte"], vert=False)
plt.axvline(lower_bound_iqr, color="red", linestyle="--", label="IQR-Grenzen")
plt.axvline(upper_bound_iqr, color="red", linestyle="--")
plt.title("IQR-Methode (Ausreißer = Punkte außerhalb der Boxplot-Whisker)")
plt.legend()

# Histogramm mit 3σ
plt.subplot(2,1,2)
plt.hist(df_test["Werte"], bins=20, alpha=0.7, edgecolor="black")
plt.axvline(lower_bound_sigma, color="green", linestyle="--", label="3σ-Grenzen")
plt.axvline(upper_bound_sigma, color="green", linestyle="--")
plt.axvline(mean, color="blue", linestyle="-", label="Mittelwert")
plt.title("3σ-Methode")
plt.legend()

plt.tight_layout()
plt.show()

(lower_bound_iqr, upper_bound_iqr, lower_bound_sigma, upper_bound_sigma)
````

## 2.2 Explorative Datenanalyse (EDA)
Explorative Datenanalyse (EDA) ist ein methodischer Ansatz zur ersten Untersuchung von Datensätzen ohne vorgegebene Hypothesen. Ziel ist es, durch statistische und visuelle Methoden Muster, Zusammenhänge, Ausreißer und Datenstrukturen zu erkennen. Dabei hilft EDA, ein besseres Verständnis der Daten zu erlangen, Hypothesen zu generieren und die Datenqualität zu überprüfen. 

EDA verwendet häufig Streudiagramme und Korrelationsanalysen, um Zusammenhänge zwischen Variablen zu untersuchen, während die deskriptive Statistik eher einfache Zusammenfassungen und Standardvisualisierungen liefert.

### 2.2.1 Streudiagramm: Preis (y-Achse) vs. numerische Spalte (x-Achse)

Ein Streudiagramm ist für die explorative Datenanalyse (EDA) sehr nützlich, weil es:

- Zusammenhänge zwischen zwei numerischen Variablen sichtbar macht

### 3.5.1 Statistische Betrachtung:
1. Was ist die Standardabweichung (σ oder std)?
 - Die Standardabweichung misst, wie stark die einzelnen Werte einer Datenmenge im Durchschnitt von deren Mittelwert (Durchschnitt) abweichen.
 
 - Kleine Standardabweichung → die Werte liegen eng um den Mittelwert herum (geringe Streuung).
 
 - Große Standardabweichung → die Werte sind weit vom Mittelwert verteilt (hohe Streuung).
 
Formel (für eine Stichprobe):



wobei
xi = einzelner Wert,
xˉ = Mittelwert,
n = Anzahl der Werte.

2. Was bedeutet „3-fache Standardabweichung“?
Wenn die Daten annähernd normalverteilt (Glockenkurve) sind, gilt die sogenannte 68–95–99,7-Regel:

 -	±1 σ → ca. 68% aller Werte liegen innerhalb einer Standardabweichung vom Mittelwert.

 -	±2 σ → ca. 95% aller Werte liegen innerhalb von zwei Standardabweichungen.

 -	±3 σ → ca. 99,7% aller Werte liegen innerhalb von drei Standardabweichungen.

3. Das heißt: die Werte außer ±3σ werden beseitigt.
Werte, die mehr als 3 Standardabweichungen vom Mittelwert entfernt sind, treten sehr selten auf (~0,3% Wahrscheinlichkeit).
In der Statistik werden diese oft als Ausreißer betrachtet.

# 4. Modeling (ML-Maschinelles Lernen) 
**Mathematische Grundlage: die Kleinste-Quadrate-Methode**
## 4.1 Der allgemeine lineare Fall
 1. model = LinearRegression()
 2. model.fit(X_train, y_train)
 
 ### 4.1.1 Ein eindimensionles Linear-Regression: 
 
 y= a*x + b; x, y: Messwerte; a: Steigerung, b: bias



 ### 4.1.2 N-dimensionles Linear-Regression: 
 
 ### die Kleinste-Quadrate-Methode(KQM)
 ### r= Xα - y; 
 m x (n+1) ; m: Anzahl der Beobachtungen(oder Anzahl der Zeilen in einer Spalte)
 
 a=(a0, a1, a2,... an), n+1-Dimension, Unbekannte;  a0: bias
 
 X: mx(n+1) dimensionale Matrix (Beobachtungen-> Xi,j: i Zeile; j: Spalte 
 
 y: m-Messwerte (Preise)
 




Ziel: die Summe durch KQM zu minimieren



Die Lösung: np.linalg.inv()




### Achtung

1. **Bedingung für eindeutige Lösung**  
   Damit die Matrix $A^T A$ stabil invertierbar ist, muss die Anzahl der **Zeilen (Beobachtungen)** stets **größer** sein als die Anzahl der **Spalten (Variablen/Features)**.  
   Ist das nicht der Fall, existiert keine eindeutige Lösung für das Gleichungssystem.

2. **Lineare Unabhängigkeit der Spalten**  
   Werden neue Spalten aus bestehenden Daten erzeugt, dürfen sie **nicht in einem proportionalen Verhältnis** zu anderen Spalten stehen. Andernfalls sind die Spalten **linear abhängig**, und die Matrix \(A^T A\) wird **singulär** (nicht invertierbar).  

   - In diesem Fall liefert `np.linalg.inv()` keine echte Inverse, sondern bestenfalls eine **Pseudo-Inverse** (z. B. über `np.linalg.pinv()`), was mathematisch problematisch ist.  
   - Beispiel:  
    - $x_1 = a \cdot x$ → ❌ **nicht erlaubt** (linear abhängig)  
    - $x_1 = \sqrt{x}$ → ✅ **erlaubt** (nicht linear abhängig)

#### Noch ein Beispiel mit Training: Datensatz (synthetisch, aber realistisch)

````python
# Datensatz (synthetisch, aber realistisch)
import torch
import torch.nn as nn
import torch.optim as optim

torch.manual_seed(0)

# künstliche Features
X = torch.randn(200, 5)

# latenter Score + Rauschen → ordinale Klassen
latent = X[:, 0] + 0.5 * X[:, 1]
y = torch.bucketize(
    latent,
    boundaries=torch.tensor([-1.0, -0.2, 0.5, 1.2])
)

print("Target distribution:", torch.bincount(y))
print("X:\n", X[0:5])
print("y:", y[0:10])

````

````python
#Ordinale Targets erzeugen (CORN-Style)
def ordinal_targets(y, num_classes):
    return torch.stack(
        [(y >= i).float() for i in range(1, num_classes)],
        dim=1
    )

num_classes = 5
y_ord = ordinal_targets(y, num_classes)
print(y_ord[:10])
````

````python
#
# CORN-Modell: CORN (Conditional Ordinal Regression for NN)
# 2 NN layers 
# num_classes=5

betw_nn_no = 16  # try with 8
class OrdinalNet(nn.Module):
    def __init__(self, in_features, num_classes):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_features, betw_nn_no),
            #nn.BatchNorm1d(betw_nn_no),  # <-- Batch Norm (nach Linear oder ReLU)
            nn.ReLU(),
            #nn.Dropout(p=0.2),           # <-- Dropout (20% der Neuronen aus)
            nn.Linear(betw_nn_no, num_classes - 1)
        )
    def forward(self, x):
        return self.net(x)  # logits
````

````python
# Training

model = OrdinalNet(5, num_classes)
optimizer = optim.Adam(model.parameters(), lr=0.01)
criterion = nn.BCEWithLogitsLoss()
epoch_n= 120
for epoch in range(1, epoch_n):
    optimizer.zero_grad()
    logits = model(X)
    loss = criterion(logits, y_ord)
    loss.backward()
    optimizer.step()

    if epoch % 20 == 0:
        print(f"Epoch {epoch:3d} | Loss {loss.item():.4f}")
````

````python
# Vorhersage → ordinale Klasse
model.eval() # 1. Layer-Verhalten auf "Vorhersage" stellen
with torch.no_grad():
    logits = model(X)
    probs = torch.sigmoid(logits)
    y_pred = (probs > 0.5).sum(dim=1)

print("True:", y[:10])
print("Pred:", y_pred[:10])
````

2. Kategorische Spalten benutzt OrdinalEncoder + numerische Spalten (standardisiert) → lineares Modell

was bedeutet "standardisiert"?

StandardScaler() ist ein Werkzeug aus scikit-learn, das numerische Features standardisiert: Es bringt alle angegebenen Spalten (Features) auf eine gemeinsame Skala, indem jeder Wert so transformiert wird:

**z = x−μσ**

**z = σx−μ**

Dabei ist **x** der Originalwert, **μ** der Mittelwert und **σ** die Standardabweichung der jeweiligen Spalte.

fit_transform() berechnet zuerst Mittelwert und Standardabweichung für jede numerische Spalte – und transformiert dann alle Werte entsprechend.

Das Ergebnis: Alle Werte der angegebenen numerischen Spalten haben danach Mittelwert=0 und Standardabweichung=1. Die Skalierung ist jetzt vergleichbar, egal ob z.B. Einkommen, Alter oder Wohnfläche vorher unterschiedlich große Zahlen waren.

## 4.4 Untersuchung von verschiedenen Verfahren
### Erklärung von r2_score(y_test, y_pred)

**R² (Bestimmtheitsmaß) - Erklärung**

####  1️⃣ Definition von R² (Bestimmtheitsmaß)

Für ein Test-Set gilt:

  R² = 1 - (Σ(y_i - ŷ_i)² / Σ(y_i - ȳ)²)

    - y_i = echte Werte  
    - ŷ_i = vorhergesagte Werte  
    - ȳ = Mittelwert der echten Werte

Interpretation:

    - R² = 1 → perfekte Vorhersage
    - R² = 0 → Modell sagt genauso gut wie Mittelwert
    - R² < 0 → Modell ist schlechter als Mittelwert

####  2️⃣ Wann wird R² negativ?

Σ(y_i - ŷ_i)² > Σ(y_i - ȳ)²

    - Das passiert, wenn dein Modell die Werte so schlecht vorhersagt, dass der Mittelwert eine bessere Vorhersage liefern würde.
    - Typische Ursachen:
    - Lineares Modell auf nichtlinearen Zusammenhängen
    - Zu wenige oder falsche Features
    - Ausreißer im Test-Set
    - Übermäßiges Rauschen in den Daten

Negatives R² ist ein Warnsignal, dass dein Modell nicht gut passt.
Es bedeutet nicht, dass der Wert falsch berechnet wurde.

````python
# training_carmodel_CV

#from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
#from sklearn.linear_model import LinearRegression
#from sklearn.metrics import mean_squared_error, r2_score

global count_train
count_train = 0

def training_carmodel_CV(
    df_cp,
    categorical_columns,
    numerical_columns,
    encoding_mode="onehot",   # "onehot" | "ordinal"
    n_splits=5,
    squared_cols=0
):
    """
    Train a linear regression model for car price prediction using StratifiedKFold.

    Parameters
    ----------
    df_cp : pandas.DataFrame
        Full dataset containing both features and the target column 'price'.

    categorical_columns : list of str
        Names of categorical feature columns.

    numerical_columns : list of str
        Names of numerical feature columns.

    encoding_mode : str, default="onehot"
        "onehot" or "ordinal"

    n_splits : int, default=5
        Number of folds for StratifiedKFold.

    squared_cols : int, default=0
        Optional nonlinear transformations: 0=None, 1=sqrt, 2=square

    Returns
    -------
    model : trained LinearRegression model on all data
    ordinal_encoders : dict or None
    onehot_encoder : fitted OneHotEncoder or None
    scaler : fitted StandardScaler
    feature_names : list of final feature names
    cv_results : DataFrame with MSE/R2 per fold
    pred_df : DataFrame of predictions on all data (final model)
    """

    global count_train
    df_cp = df_cp.copy()
    X = df_cp.drop(columns=["price"])
    y = df_cp["price"]

    # Stratified bins für Regression
    y_bins = pd.qcut(y, q=5, duplicates='drop')

    # Stratified KFold
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

    fold_results = []

    for fold, (train_idx, test_idx) in enumerate(skf.split(X, y_bins), 1):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

        # --------------------------------------------------
        # Categorical Encoding
        # --------------------------------------------------
        onehot_encoder = None
        ordinal_encoders = None

        if encoding_mode == "onehot":
            onehot_encoder = OneHotEncoder(
                drop="first",
                handle_unknown="ignore",
                sparse_output=False
            )

            X_train_cat = onehot_encoder.fit_transform(
                X_train[categorical_columns]
            )
            X_test_cat = onehot_encoder.transform(
                X_test[categorical_columns]
            )

            cat_feature_names = onehot_encoder.get_feature_names_out(
                categorical_columns
            )

            X_train_cat = pd.DataFrame(
                X_train_cat,
                columns=cat_feature_names,
                index=X_train.index
            )

            X_test_cat = pd.DataFrame(
                X_test_cat,
                columns=cat_feature_names,
                index=X_test.index
            )

        elif encoding_mode == "ordinal":
            ordinal_encoders = {}
            X_train_cat = pd.DataFrame(index=X_train.index)
            X_test_cat = pd.DataFrame(index=X_test.index)

            for col in categorical_columns:
                oe = OrdinalEncoder(
                    handle_unknown="use_encoded_value",
                    unknown_value=-1
                )

                X_train_cat[col] = oe.fit_transform(X_train[[col]]).ravel()
                X_test_cat[col] = oe.transform(X_test[[col]]).ravel()
                ordinal_encoders[col] = oe

        else:
            raise ValueError("encoding_mode must be 'onehot' or 'ordinal'")

        # --------------------------------------------------
        # Numerical Features + Feature Engineering
        # --------------------------------------------------
        X_train_num = X_train[numerical_columns].copy()
        X_test_num = X_test[numerical_columns].copy()

        # Engineered feature (power_to_weight_ratio)
        X_train_num["power_to_weight_ratio"] = (
            X_train["horsepower"] / X_train["curbweight"].replace(0, np.nan)
        ).fillna(0)
        X_test_num["power_to_weight_ratio"] = (
            X_test["horsepower"] / X_test["curbweight"].replace(0, np.nan)
        ).fillna(0)

        # Optional nonlinear transformations
        for col in numerical_columns:
            if squared_cols == 2:
                X_train_num[f"{col}_squared"] = X_train[col] ** 2
                X_test_num[f"{col}_squared"] = X_test[col] ** 2
            elif squared_cols == 1:
                X_train_num[f"{col}_squared"] = np.sqrt(np.abs(X_train[col]))
                X_test_num[f"{col}_squared"] = np.sqrt(np.abs(X_test[col]))

        # Scaling
        scaler = StandardScaler()
        X_train_num = pd.DataFrame(
            scaler.fit_transform(X_train_num),
            columns=X_train_num.columns,
            index=X_train_num.index
        )
        X_test_num = pd.DataFrame(
            scaler.transform(X_test_num),
            columns=X_test_num.columns,
            index=X_test_num.index
        )

        # Final Design Matrices
        X_train_final = pd.concat([X_train_num, X_train_cat], axis=1)
        X_test_final = pd.concat([X_test_num, X_test_cat], axis=1)

        feature_names = X_train_final.columns

        # Model Training
        model = LinearRegression()
        model.fit(X_train_final, y_train)

        # Evaluation
        y_pred = model.predict(X_test_final)
        mse = mean_squared_error(y_test, y_pred)
        r2 = r2_score(y_test, y_pred)

        fold_results.append({
            "fold": fold,
            "mse": mse,
            "r2": r2
        })

        count_train += 1

    cv_results = pd.DataFrame(fold_results)

    # --------------------------------------------------
    # Final Model auf allen Daten
    # --------------------------------------------------
    # Encoding + Scaling auf kompletter Datenbasis
    if encoding_mode == "onehot":
        onehot_encoder = OneHotEncoder(
            drop="first",
            handle_unknown="ignore",
            sparse_output=False
        )
        X_cat_all = onehot_encoder.fit_transform(X[categorical_columns])
        cat_feature_names = onehot_encoder.get_feature_names_out(categorical_columns)
        X_cat_all = pd.DataFrame(X_cat_all, columns=cat_feature_names, index=X.index)
    else:
        ordinal_encoders = {}
        X_cat_all = pd.DataFrame(index=X.index)
        for col in categorical_columns:
            oe = OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)
            X_cat_all[col] = oe.fit_transform(X[[col]]).ravel()
            ordinal_encoders[col] = oe

    X_num_all = X[numerical_columns].copy()
    X_num_all["power_to_weight_ratio"] = (X["horsepower"] / X["curbweight"].replace(0, np.nan)).fillna(0)
    for col in numerical_columns:
        if squared_cols == 2:
            X_num_all[f"{col}_squared"] = X[col] ** 2
        elif squared_cols == 1:
            X_num_all[f"{col}_squared"] = np.sqrt(np.abs(X[col]))

    scaler = StandardScaler()
    X_num_all = pd.DataFrame(scaler.fit_transform(X_num_all), columns=X_num_all.columns, index=X.index)

    X_final = pd.concat([X_num_all, X_cat_all], axis=1)
    feature_names = X_final.columns

    final_model = LinearRegression()
    final_model.fit(X_final, y)

    # Predictions auf allen Daten
    y_pred_all = final_model.predict(X_final)
    pred_df = pd.DataFrame({
        "Actual Value": y,
        "Predicted Value": y_pred_all,
        "Difference": y - y_pred_all,
        "rel%": np.abs(y - y_pred_all) / y * 100
    })

    return final_model, ordinal_encoders, onehot_encoder, scaler, feature_names, cv_results, pred_df
````

````python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

from sklearn.linear_model import (
    LinearRegression, Ridge, Lasso, ElasticNet, HuberRegressor
)
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor


from sklearn.base import BaseEstimator, TransformerMixin

# Custom Feature Engineering (Pipeline-kompatibel!)


class PowerToWeightAdder(BaseEstimator, TransformerMixin):
    def __init__(self, hp_col="horsepower", weight_col="curbweight"):
        self.hp_col = hp_col
        self.weight_col = weight_col

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = X.copy()
        ratio = X[self.hp_col] / X[self.weight_col].replace(0, np.nan)
        X["power_to_weight_ratio"] = ratio.fillna(0)
        return X

    def get_feature_names_out(self, input_features=None):
        return list(input_features) + ["power_to_weight_ratio"]



# Pipeline Builder (zentraler Baustein)
def build_pipeline(categorical_columns, numerical_columns, model):
    num_pipeline = Pipeline(steps=[
        ("feat", PowerToWeightAdder()),
        ("scaler", StandardScaler())
    ])

    cat_pipeline = Pipeline(steps=[
        ("onehot", OneHotEncoder(
            drop="first",
            handle_unknown="ignore",
            sparse_output=False
        ))
    ])

    preprocessor = ColumnTransformer(
        transformers=[
            ("num", num_pipeline, numerical_columns),
            ("cat", cat_pipeline, categorical_columns)
        ],
        remainder="drop"
    )

    pipe = Pipeline(steps=[
        ("prep", preprocessor),
        ("model", model)
    ])

    return pipe

# GridSearch-Setup (alle Modelle!)
def run_gridsearch(
    df,
    categorical_columns,
    numerical_columns,
    test_size=0.2 # 0.2
):
    X = df.drop(columns=["price"])
    y = df["price"]

    # stratified regression split
    y_bins = pd.qcut(y, q=5, duplicates="drop")

    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=test_size,
        random_state=42,
        stratify=y_bins
    )

    pipe = build_pipeline(
        categorical_columns,
        numerical_columns,
        model=LinearRegression()  # placeholder
    )
    param_grid = [
        {
            "model": [LinearRegression()]
        },
        {
            "model": [Ridge()],
            "model__alpha": [0.1, 1.0, 10.0]
        },
        {
            "model": [ElasticNet(max_iter=10_000)],
            "model__alpha": [0.01, 0.1],
            "model__l1_ratio": [0.2, 0.5, 0.8]
        },
        {
            "model": [HuberRegressor()],
            "model__epsilon": [1.2, 1.35, 1.5]
        },

        {
            "model": [RandomForestRegressor(random_state=42)],
            "model__n_estimators": [200, 400],
            "model__max_depth": [10, 15]
        },
        {
            "model": [GradientBoostingRegressor(random_state=42)],
            "model__n_estimators": [200, 400],
            "model__learning_rate": [0.03, 0.05],
            "model__max_depth": [2, 3]
        }
    ]

    gs = GridSearchCV(
        pipe,
        param_grid,
        scoring="r2",
        cv=5,
        n_jobs=-1,
        verbose=2
    )

    gs.fit(X_train, y_train)

    best_model = gs.best_estimator_
    # Zugriff auf Preprocessor
    preprocessor = best_model.named_steps["prep"]

    # Numerische Pipeline
    num_pipeline = preprocessor.named_transformers_["num"]
    scaler = num_pipeline.named_steps["scaler"]

    # Kategorische Pipeline
    cat_pipeline = preprocessor.named_transformers_["cat"]
    onehot = cat_pipeline.named_steps["onehot"]

    feature_names = preprocessor.get_feature_names_out()

    y_pred = best_model.predict(X_test)
    
    cv_results = gs.cv_results_
    
    # Erstelle DataFrame für einfache Analyse
    results_df = pd.DataFrame(cv_results) 
    
    # Identifiziere Modelltyp über 'param_model' Spalte (der Estimator-Name)
    results_df['model_type'] = results_df['param_model'].apply(lambda x: type(x).__name__)
    
    # Bestes R² pro Modelltyp (mean_test_score ist der CV-R²)
    best_per_model = results_df.groupby('model_type')['mean_test_score'].max().sort_values(ascending=False)
    
    # Mit besten Parametern
    best_per_model_detailed = results_df.loc[results_df.groupby('model_type')['mean_test_score'].idxmax()]

    pred_df = pd.DataFrame({
        "Actual Value": y_test,
        "Predicted Value": y_pred,
        "Difference": y_test - y_pred,
        "rel%": np.abs(y_test - y_pred) / y_test * 100
    })

    return {
        "best_model": best_model,
        "best_model_name": gs.best_params_["model"].__class__.__name__,
        "onehot": onehot,
        "scaler": scaler,
        "feature_names": feature_names,
        "best_params": gs.best_params_,
        "cv_best_score": gs.best_score_,
        "test_r2": r2_score(y_test, y_pred),
        "test_mse": mean_squared_error(y_test, y_pred),
        "best_r2_per_model": best_per_model,  # Serie mit besten CV-R² pro Modell
        "best_per_model_detailed": best_per_model_detailed,  # komplette Zeile pro bestem Modell
        "all_results": results_df,  # alles für tiefe Analyse
        "pred_df":pred_df
    }
    
````

### 4.4.5  training_carmodel() für die Anwendung 



````python
global count_train

count_train=0
def training_carmodel(
    df_cp,
    categorical_columns,
    numerical_columns,
    encoding_mode="onehot",   # "onehot" | "ordinal"
    test_daten_size=0.2,
    squared_cols=0
):
    global count_train
    
    """
    Train a linear regression model for car price prediction.

    Parameters
    ----------
    df_cp : pandas.DataFrame
        Full dataset containing both features and the target column
        named 'price'. The DataFrame must include all categorical
        and numerical columns specified below.

    categorical_columns : list of str
        Names of categorical feature columns.
        These columns will be encoded using either One-Hot Encoding
        or Ordinal Encoding depending on `encoding_mode`.

    numerical_columns : list of str
        Names of numerical feature columns.
        These features will be scaled using StandardScaler and may
        optionally receive nonlinear transformations.

    encoding_mode : str, default="onehot"
        Encoding strategy for categorical features:
        - "onehot": One-Hot Encoding using sklearn's OneHotEncoder
                    (recommended for linear models).
        - "ordinal": Ordinal Encoding using OrdinalEncoder with
                     unknown categories mapped to -1
                     (intentionally information-losing; for experiments).

    test_daten_size : float, default=0.2
        Proportion of the dataset to include in the test split.
        Must be a value between 0 and 1.

    squared_cols : int, default=0
        Controls optional nonlinear transformations of numerical features:
        - 0 : No transformation
        - 1 : Square root transformation (sqrt(|x|))
        - 2 : Squared transformation (x²)

    Returns
    -------
    model : sklearn.linear_model.LinearRegression
        Trained linear regression model.

    ordinal_encoders : dict or None
        Dictionary mapping column names to fitted OrdinalEncoder objects.
        Returned only if `encoding_mode="ordinal"`, otherwise None.

    onehot_encoder : sklearn.preprocessing.OneHotEncoder or None
        Fitted OneHotEncoder used during training.
        Returned only if `encoding_mode="onehot"`, otherwise None.

    scaler : sklearn.preprocessing.StandardScaler
        Scaler fitted on the numerical training features.

    feature_names : pandas.Index
        Ordered list of feature names used during training.
        Required to ensure consistent feature alignment during prediction.

    mse : float
        Mean Squared Error evaluated on the test set.

    r2_square : float
        Coefficient of determination (R² score) evaluated on the test set.

    pred_df : pandas.DataFrame
        DataFrame comparing actual target values and model predictions,
        including absolute and relative errors.
    """

    # --------------------------------------------------
    # Defensive copy
    # --------------------------------------------------
    df_cp = df_cp.copy()

    # --------------------------------------------------
    # Train / Test Split (NO leakage!)
    # --------------------------------------------------
    X = df_cp.drop(columns=["price"])
    y = df_cp["price"]

    """ without stratify=y_bins
    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=test_daten_size,
        random_state=42
    )
    """
    
    # Stratified bins für Regression
    y_bins = pd.qcut(y, q=5, duplicates='drop')

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=test_daten_size,
        random_state=42,
        stratify=y_bins
    )



    # --------------------------------------------------
    # Categorical Encoding
    # --------------------------------------------------
    onehot_encoder = None
    ordinal_encoders = None

    if encoding_mode == "onehot":
        onehot_encoder = OneHotEncoder(
            drop="first",
            handle_unknown="ignore",
            sparse_output=False
        )

        X_train_cat = onehot_encoder.fit_transform(
            X_train[categorical_columns]
        )
        X_test_cat = onehot_encoder.transform(
            X_test[categorical_columns]
        )

        cat_feature_names = onehot_encoder.get_feature_names_out(
            categorical_columns
        )

        X_train_cat = pd.DataFrame(
            X_train_cat,
            columns=cat_feature_names,
            index=X_train.index
        )

        X_test_cat = pd.DataFrame(
            X_test_cat,
            columns=cat_feature_names,
            index=X_test.index
        )

    elif encoding_mode == "ordinal":
        ordinal_encoders = {}

        X_train_cat = pd.DataFrame(index=X_train.index)
        X_test_cat = pd.DataFrame(index=X_test.index)

        for col in categorical_columns:
            oe = OrdinalEncoder(
                handle_unknown="use_encoded_value",
                unknown_value=-1
            )

            X_train_cat[col] = oe.fit_transform(
                X_train[[col]]
            ).ravel()

            X_test_cat[col] = oe.transform(
                X_test[[col]]
            ).ravel()

            ordinal_encoders[col] = oe

    else:
        raise ValueError("encoding_mode must be 'onehot' or 'ordinal'")

    # --------------------------------------------------
    # Numerical Features + Feature Engineering
    # --------------------------------------------------
    X_train_num = X_train[numerical_columns].copy()
    X_test_num = X_test[numerical_columns].copy()

    # Engineered feature (safe division)
    X_train_num["power_to_weight_ratio"] = (
        X_train["horsepower"] /
        X_train["curbweight"].replace(0, np.nan)
    ).fillna(0)

    X_test_num["power_to_weight_ratio"] = (
        X_test["horsepower"] /
        X_test["curbweight"].replace(0, np.nan)
    ).fillna(0)

    # Optional nonlinear transformations
    for col in numerical_columns:
        if squared_cols == 2:
            X_train_num[f"{col}_squared"] = X_train[col] ** 2
            X_test_num[f"{col}_squared"] = X_test[col] ** 2

        elif squared_cols == 1:
            X_train_num[f"{col}_squared"] = np.sqrt(np.abs(X_train[col]))
            X_test_num[f"{col}_squared"] = np.sqrt(np.abs(X_test[col]))

    # --------------------------------------------------
    # Scaling (SAFE Pandas handling)
    # --------------------------------------------------
    scaler = StandardScaler()

    X_train_num = pd.DataFrame(
        scaler.fit_transform(X_train_num),
        columns=X_train_num.columns,
        index=X_train_num.index
    )

    X_test_num = pd.DataFrame(
        scaler.transform(X_test_num),
        columns=X_test_num.columns,
        index=X_test_num.index
    )

    print("X_train:cat, num, X_Test:cat, num", X_train_cat.shape, X_train_num.shape, X_test_cat.shape, X_test_num.shape)

    count_train=count_train+1
    print("count of the trainig: ", count_train)
    
    # --------------------------------------------------
    # Final Design Matrices
    # --------------------------------------------------
    X_train_final = pd.concat(
        [X_train_num, X_train_cat],
        axis=1
    )

    X_test_final = pd.concat(
        [X_test_num, X_test_cat],
        axis=1
    )

    # 🔒 Save feature order (CRITICAL!)
    feature_names = X_train_final.columns

    # --------------------------------------------------
    # Model Training
    # HuberRegressor(epsilon=1.35): showed as the best model
    # --------------------------------------------------
    #model = LinearRegression()    
    #model = Ridge(alpha=1.0)
    #model= Lasso(alpha=0.01, max_iter=10_000) 
    #model = ElasticNet(alpha=0.01, l1_ratio=0.5, max_iter=10_000)
    model = HuberRegressor(epsilon=1.35)
    #model = RandomForestRegressor(n_estimators=300,max_depth=15,random_state=42,n_jobs=-1)  

    #model = GradientBoostingRegressor( n_estimators=300,learning_rate=0.05,max_depth=3,random_state=42)


    
    model.fit(X_train_final, y_train)

    # --------------------------------------------------
    # Evaluation
    # --------------------------------------------------
    y_pred = model.predict(X_test_final)

    mse = mean_squared_error(y_test, y_pred)
    r2_square = r2_score(y_test, y_pred)

    #print("y_test:\n", y_test)
    pred_df = pd.DataFrame({
        "Actual Value": y_test,
        "Predicted Value": y_pred,
        "Difference": y_test - y_pred,
        "rel%": np.abs(y_test - y_pred) / y_test * 100
    })


    return (
        model,
        ordinal_encoders,
        onehot_encoder,
        scaler,
        feature_names,
        mse,
        r2_square,
        pred_df
    )
````

### 4.4.6 def predict_price()

````python
def predict_price(
    df_test,
    model,
    ordinal_encoders,
    onehot_encoder,
    scaler,
    feature_names,
    categorical_columns,
    numerical_columns,
    squared_cols=0
):
    """
    Predict car prices using a trained regression model.

    Parameters
    ----------
    df_test : pandas.DataFrame
        New, unseen data containing the same feature columns
        as used during training (excluding the target column 'price').

    model : sklearn.linear_model.LinearRegression
        Trained regression model returned by `training_carmodel`.

    ordinal_encoders : dict or None
        Dictionary of fitted OrdinalEncoders used during training.
        Must be provided if ordinal encoding was used, otherwise None.

    onehot_encoder : sklearn.preprocessing.OneHotEncoder or None
        Fitted OneHotEncoder from training.
        Must be provided if one-hot encoding was used, otherwise None.

    scaler : sklearn.preprocessing.StandardScaler
        Scaler fitted on the numerical training features.

    feature_names : pandas.Index
        Ordered list of feature names used during training.
        Ensures correct feature alignment during prediction.

    categorical_columns : list of str
        Names of categorical feature columns.

    numerical_columns : list of str
        Names of numerical feature columns.

    squared_cols : int, default=0
        Must match the transformation mode used during training:
        - 0 : No transformation
        - 1 : Square root transformation
        - 2 : Squared transformation

    Returns
    -------
    y_pred : numpy.ndarray
        Predicted car prices for each row in `df_test`.
    """

    df_test = df_test.copy()
    # --------------------------------------------------
    # Encode categorical features
    # --------------------------------------------------
    if onehot_encoder is not None:
        X_cat = onehot_encoder.transform(
            df_test[categorical_columns]
        )

        cat_feature_names = onehot_encoder.get_feature_names_out(
            categorical_columns
        )

        X_cat = pd.DataFrame(
            X_cat,
            columns=cat_feature_names,
            index=df_test.index
        )
    else:
        X_cat = pd.DataFrame(index=df_test.index)
        for col in categorical_columns:
            X_cat[col] = ordinal_encoders[col].transform(
                df_test[[col]]
            ).ravel()

    # --------------------------------------------------
    # Numerical features + feature engineering
    # --------------------------------------------------
    X_num = df_test[numerical_columns].copy()

    X_num["power_to_weight_ratio"] = (
        df_test["horsepower"] /
        df_test["curbweight"].replace(0, np.nan)
    ).fillna(0)

    for col in numerical_columns:
        if squared_cols == 2:
            X_num[f"{col}_squared"] = df_test[col] ** 2
        elif squared_cols == 1:
            X_num[f"{col}_squared"] = np.sqrt(np.abs(df_test[col]))

    # --------------------------------------------------
    # Scaling (SAFE)
    # --------------------------------------------------
    X_num = pd.DataFrame(
        scaler.transform(X_num),
        columns=X_num.columns,
        index=X_num.index
    )

    # --------------------------------------------------
    # Final feature matrix (ORDER MATTERS!)
    # --------------------------------------------------
    X_final = pd.concat([X_num, X_cat], axis=1)
    #print("predict_price:  X_final:", X_final)
    #print(feature_names)
    
    X_final = X_final[feature_names]

    # --------------------------------------------------
    # Prediction
    # --------------------------------------------------
    y_pred = model.predict(X_final)

    return y_pred
````

# Conclusion and Outlook

1. Project Structure Based on CRISP-DM

This study was conducted entirely in accordance with the CRISP-DM methodology. All phases, including descriptive analysis and exploratory data analysis (EDA), were systematically executed to ensure a structured and reproducible workflow.

2. Initial Data Overview

After loading the dataset, standard inspection methods such as df.info(), df.describe(), df.nunique(), df.isna().sum(), and df.head() were applied to obtain an initial understanding of data types, distributions, missing values, and overall structure.

3. Descriptive Statistical Analysis

The dataset was subsequently examined using descriptive statistical measures to gain deeper insights into central tendencies, dispersion, and distributional characteristics of the variables.

4. Exploratory Data Analysis (EDA)

Exploratory data analysis was employed to identify patterns, correlations, and potential dependencies between variables. These insights provided guidance for feature selection and preprocessing decisions in later stages of the analysis.

5. Data Preprocessing

Following the statistical analysis, data preprocessing was performed. Given the relatively small size of the dataset (205 observations and 26 features), preprocessing was deliberately kept minimal but targeted. The steps included:

Decomposition of compound columns

Transformation of categorical variables into numerical representations

Identification and treatment of outliers

These steps aimed to preserve as much information as possible while ensuring model compatibility and numerical stability.

6. Machine Learning

The core contribution of this work lies in the systematic machine learning analysis.

A detailed comparison of label (ordinal) encoding and one-hot encoding was conducted, with particular attention to their theoretical implications and practical impact on model performance. Since most categorical variables in the dataset are nominal in nature, ordinal encoding introduces artificial orderings that violate the assumptions of linear models and may distort distance-based interpretations.

In the project dataset, the categorical features do not possess an inherent order, making label or ordinal encoding unsuitable for representing these variables. Consequently, the use of ordinal encodings led to inferior model performance compared to one-hot encoding.

Therefore, in this work, label and ordinal encoding were applied only for experimental and comparison purposes, while one-hot encoding was selected as the primary encoding strategy for categorical variables.


To mitigate the high dimensionality resulting from one-hot encoding, feature engineering techniques were applied to reduce the number of categorical features while retaining relevant information.

Multiple regression models were evaluated, and the influence of outlier treatment was explicitly examined. The results demonstrate that removing two extreme outliers led to improved model robustness and predictive performance.

Model evaluation was carried out using both holdout validation and k-fold cross-validation, enabling a reliable assessment of generalization performance. This methodological comparison highlights the stability of the results despite the limited dataset size.

Overall, the experiments show that RandomForestRegressor and GradientBoostingRegressor deliver stable performance on the small dataset used in this study.
Due to their robustness with respect to non-linearities and feature interactions, these models were chosen for the final prediction task.
    
7. Dashboard: Model Training

Two interactive dashboards were developed as part of the project. The model training dashboard allows users to select and train models within a specified price segment.

A practical limitation arises from the dataset size: the selected price range must be sufficiently wide to ensure an adequate number of samples for statistically meaningful model training.

8. Dashboard: Price Prediction

After training, price prediction can be performed directly through the second dashboard:

The user is able to enter a reference price, upon which the system searches for the closest matching item in the original dataset. This item and its corresponding price are displayed in the dashboard. The displayed price is then used as a reference value to compare against the predicted price when the user adjusts the vehicle parameters.

In this implementation, only two parameters are used to modify vehicle characteristics. In a real production setting, additional parameters would be included.

The trained model predicts the price for the modified vehicle configuration.

Empirical tests indicate that the prediction framework performs reliably even with a dataset of only 205 observations. With larger datasets, further improvements in model accuracy and robustness can be expected.
