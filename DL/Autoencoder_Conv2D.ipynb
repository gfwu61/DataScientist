# 1 Autoencoder mit Conv2D:  Denoising mit Conv2D-Autoencoder
## 1.1 Aufgabestellung 

Im Rahmen dieser Arbeit werden synthetische Bilddaten generiert, die einzelne Buchstaben darstellen. Die erzeugten Bilder werden gezielt mit additivem Rauschen überlagert, um realistische Störungen zu simulieren.

Ziel ist die Entwicklung und Anwendung eines Autoencoders, der in der Lage ist, die verrauschten Eingabebilder zu komprimieren und dabei relevante Merkmale zu extrahieren. Durch den Rekonstruktionsprozess sollen die ursprünglichen Buchstabenbilder möglichst rauschfrei wiederhergestellt werden.

Die Hauptaufgabe besteht darin, den Autoencoder so zu trainieren, dass er die zugrundeliegenden Strukturen der Buchstaben erkennt und das Rauschen effektiv unterdrückt. Bei der Rekonstruktion der Testdaten soll überprüft werden, in welchem Maße das Modell in der Lage ist, das Rauschen zu eliminieren und die Bildqualität zu verbessern.

## 1.4 Entwurf von Encoder und Decoder (Symmetrie der Struktur)



Der Hauptzweck ist das effiziente Erlernen und die Rekonstruktion einer komprimierten und informativen Darstellung von BildDaten.

### 1.4.1 Zweck des Encoders
Der Encoder nimmt die Eingangsbilder und reduziert sie auf einen niedrigdimensionalen Vektor, die latente Repräsentation. Das Ziel ist, die wichtigsten Merkmale des Originals zu komprimieren und dabei unwichtige Details und Rauschen zu verwerfen.

### 1.4.2 Zweck des Decoders
Der Decoder nimmt diese komprimierte Darstellung und versucht, daraus das ursprüngliche Bilder möglichst genau zu rekonstruieren. Das Netzwerk lernt dabei, welche Merkmale für die Rekonstruktion unverzichtbar sind und wie unwichtige Informationen ignoriert werden

### 1.4.3 Workflow: 
1. encoder nimmt Bilder und gibt Features aus. -> Feature (komprimierte Bilder) kann gespeichert

2. decoder nimmt Features und gibt Bilder aus. -> Aus Features( Datei für die Features) können Bilder erzeugt werden (Bilder rekonstruieren)

3. autoencoder verbindet beides — nimmt also Bilder rein und gibt Bilder raus.

1. **Rekonstruktion von X_clean_train**: Noisen sollen weitergehend unterdrückt(gelöscht) werden.

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from PIL import Image, ImageDraw, ImageFont
from tensorflow.keras import layers, models

# ==============================
# 1. Daten erzeugen (Buchstabenbilder)
# ==============================
n_samples = 400       # Anzahl Bilder: 200
test_size= int(n_samples*0.15)  # 60,  15% von Daten als Testdaten
print(test_size)
img_size = (128, 128)
chars = list("ABCDEFGHIKLMNOPQRSTUVWXYZ")  # Bilder mit Buchstabe
np.random.seed(42)

def create_char_image(char, size=(128, 128)):
    img = Image.new("RGB", size, color=(255, 255, 255)) # erzeugt ein weißes RGB-Bild.
    draw = ImageDraw.Draw(img)                          # erzeugt ein Zeichenobjekt, mit dem wir auf img malen können.
    # font definition
    try:
        font = ImageFont.truetype("arial.ttf", 80)     # Versucht, die Schriftart Arial mit Größe 80 zu laden.
    except:
        font = ImageFont.load_default()                # Wenn Arial nicht gefunden wird, wird die Standard-Schriftart von Pillow verwendet.

    # Kompatibilität Pillow >=10 / <10
    try:
        bbox = draw.textbbox((0, 0), char, font=font)  # draw.textbbox (Pillow ≥ 10) liefert die Bounding-Box (x0, y0, x1, y1) des Textes.
        text_w = bbox[2] - bbox[0]                     # w: x1-x0; h= y1-y0
        text_h = bbox[3] - bbox[1]
    except AttributeError:
        text_w, text_h = draw.textsize(char, font=font)

    pos = ((size[0] - text_w) // 2, (size[1] - text_h) // 2)  # Berechnet die x- und y-Position, sodass der Text zentriert im Bild ist.
    draw.text(pos, char, fill=(0, 0, 0), font=font)
    return np.array(img)


# ähnlich wie bei r2_score, pixelweise berechnen
# für Anwendung: autoencoder.compile(optimizer='adam', loss='mse', metrics=[r2_metric])
def r2_metric(y_true, y_pred):
    ss_res = tf.reduce_sum(tf.square(y_true - y_pred))
    ss_tot = tf.reduce_sum(tf.square(y_true - tf.reduce_mean(y_true)))
    return 1 - ss_res / (ss_tot + tf.keras.backend.epsilon())

#autoencoder.compile(optimizer='adam', loss='mse', metrics=[r2_metric])

# Bilder erzeugen:
X_clean = []
for _ in range(n_samples):
    char = np.random.choice(chars)
    img = create_char_image(char, img_size)
    X_clean.append(img)
    
print("### img.shape:", img.shape) #  img.shape: (128, 128, 3)

X_clean = np.array(X_clean).astype("float32") / 255.0
X_clean_train= X_clean[0:n_samples-test_size]
X_clean_test= X_clean[-test_size:]

print("X_clean.shape:",  X_clean.shape)   # X_clean.shape: (200, 128, 128, 3)
print("X_clean_train.shape:",  X_clean_train.shape)
print("X_clean_Test.shape:",  X_clean_test.shape)

# ==============================
# 2️ Rauschen hinzufügen
# ==============================
noise_factor = 0.3
X_noisy = X_clean + noise_factor * np.random.randn(*X_clean.shape)
X_noisy = np.clip(X_noisy, 0., 1.) # Werte < 0 werden auf 0 gesetzt. Werte > 1 werden auf 1 gesetzt.

X_noisy_train= X_noisy[0:n_samples-test_size]
X_noisy_test= X_noisy[-test_size:]
print("X_noisy_train.shape:",  X_noisy_train.shape)
print("X_noisy_Test.shape:",  X_noisy_test.shape)


# ==============================
# 3️. Encoder
# ==============================
encoder = models.Sequential([
    layers.Input(shape=(128, 128, 3)),                             # Bild Größe: 128x128 , 3 Kanäle
    layers.Conv2D(32, (3, 3), activation='relu', padding='same'),  # 32 Filter-> 32 Kanäle
    layers.MaxPooling2D((2, 2), padding='same'),                   # Bild Größe: 128x128 => 64x64
    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),  # 64 Filter: 32 Kanäle -> 64 Kanäle
    layers.MaxPooling2D((2, 2), padding='same'),                   # Bild Größe:64x64 =>32x32
    layers.Flatten(),                                              # Bild Größe: 32x32 x 64-> 1D: 65536 
    layers.Dense(128, activation='relu')                           # Bild Größe:1D: 65536 -> 128 Neurons
])
# Kompressionsfaktor ≈ (128×128×3) / 128 = 384:1
# ==============================
# 4️. Decoder
# ==============================
decoder = models.Sequential([
    layers.Input(shape=(128,)),                          # nimmt den 128-D-Vektor als Input
    layers.Dense(32 * 32 * 64, activation='relu'),       # bringt auf die Dimension zurück, 
                                                         # die nach Flatten kam: 32x32 x 64-> 1D: 65536
    layers.Reshape((32, 32, 64)),                        # Bildgröße: 32x32,  64 Kanäle: 
                                                         #Diese Featuremaps sind noch relativ abstrakt / grob
                                                         # sie enthalten nicht direkt die Bilddetails.   
    # Conv2DTranspose() dient zur Rekonstruktion feiner Strukturen und nichtlinearer Transformation.
    layers.Conv2DTranspose(64, (3, 3), activation='relu', padding='same'), # Bildgröße: 32x32,  64 Kanäle:
    layers.UpSampling2D((2, 2)),                         # Upsampling, Bildgröße: 64x64,  64 Kanäle:
    layers.Conv2DTranspose(32, (3, 3), activation='relu', padding='same'), # Bildgröße: 64x64, 32 Kanäle
    layers.UpSampling2D((2, 2)),                         # Upsampling: Bildgröße:128x128, 32 Kanäle
    
    # Rekonstruktion auf 3 Kanäle; Sigmoid für [0,1]-Normalisierung, Bildgröße:128x128, 3 Kanäle
    layers.Conv2DTranspose(3, (3, 3), activation='sigmoid', padding='same')  
])


# ==============================
# 5️.  Autoencoder kombinieren: Functional API von Keras.
# ==============================
input_img = layers.Input(shape=(128, 128, 3))
encoded = encoder(input_img)
decoded = decoder(encoded)
autoencoder = models.Model(input_img, decoded)
#autoencoder.compile(optimizer='adam', loss='mse')
autoencoder.compile(optimizer='adam', loss='mse', metrics=[r2_metric])

# ==============================
# 6️. Training
# Kombination:
# 1. X_clean, X_clean: lernt die Strukture, nicht für denoising  vorgesehen, implizite teilweise denoising
# 2. X_noisy, X_clean: Denoising Autoencoder
# 3.  X_noisy, X_noisy: implizite teilweise denoising 
# weil Rauschen zufällig ist und nicht strukturiert, es hat also keine konsistente Repräsentation, die der Encoder lernen kann.
# ==============================
history = autoencoder.fit(
    X_noisy_train, X_noisy_train , # X_clean_train,            # X_clean,
    #X_clean_train, X_clean_train , # X_clean_train,            # X_clean,
    epochs=15,                   #10,15,20,  25, 30
    batch_size=8,
    shuffle=True,
    validation_split=0.2,          # 20 % der Daten für Validierung
    verbose=1
)

autoencoder.summary()

#encoder.save("encoder_part.keras")
#decoder.save("decoder_part.keras")
#encoder_loaded = load_model("encoder_part.keras")
#decoder_loaded = load_model("decoder_part.keras")

# autoencoder.save("autoencoder_model.keras")
# autoencoder_loaded = load_model("autoencoder_model.keras")
# autoencoder.summary()

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoche')
plt.ylabel('Loss')
plt.legend()
plt.show()


# ==============================
# 7️.  Rekonstruktion testen
# ==============================
decoded_imgs_train = autoencoder.predict(X_noisy_train)
decoded_imgs_test = autoencoder.predict(X_noisy_test)

# decoded_imgs = autoencoder.predict(X_clean)

# ==============================
# 8️.  Anzeige Original vs. Rauschen vs. Rekonstruktion
# ==============================

print("\n======================================================================")
print(" Denoising für Traindaten\n")

n = 6
plt.figure(figsize=(12, 6))
for i in range(n):
    # Original
    ax = plt.subplot(3, n, i + 1)
    plt.imshow(X_clean_train[i])
    plt.title("Original")
    plt.axis("off")

    # Noisy
    ax = plt.subplot(3, n, i + 1 + n)
    plt.imshow(X_noisy_train[i])
    plt.title("Rauschen")
    plt.axis("off")

    # Rekonstruiert
    ax = plt.subplot(3, n, i + 1 + 2*n)
    plt.imshow(decoded_imgs_train[i])
    plt.title("Rekonstruiert")
    plt.axis("off")

plt.tight_layout()
plt.show()
print("\n======================================================================")
print(" Denoising für Testdaten\n")

n = 6
plt.figure(figsize=(12, 6))
for i in range(n):
    # Original
    ax = plt.subplot(3, n, i + 1)
    plt.imshow(X_clean_test[i])
    plt.title("Original")
    plt.axis("off")

    # Noisy
    ax = plt.subplot(3, n, i + 1 + n)
    plt.imshow(X_noisy_test[i])
    plt.title("Rauschen")
    plt.axis("off")

    # Rekonstruiert
    ax = plt.subplot(3, n, i + 1 + 2*n)
    plt.imshow(decoded_imgs_test[i])
    plt.title("Rekonstruiert")
    plt.axis("off")

plt.tight_layout()
plt.show()

```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model
from tensorflow.keras import optimizers


# ======================
# 1️ Dummy-Bilddaten
# ======================
# 1000 RGB-Bilder, 128x128
X = np.random.rand(1000, 128, 128, 3).astype("float32")

# Normalisierung (optional, da schon [0,1])
X = X / np.max(X)

# ======================
# 2️ Encoder
# ======================
encoder = models.Sequential([
    layers.Input(shape=(128, 128, 3)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),
    layers.Dense(128, activation='relu')
])

# ======================
# 3️ Decoder
# ======================
decoder = models.Sequential([
    layers.Input(shape=(128,)),
    layers.Dense(61*61*64, activation='relu'),
    layers.Reshape((61, 61, 64)),
    layers.Conv2DTranspose(32, (3,3), activation='relu'),
    layers.UpSampling2D((2, 2)),
    layers.Conv2DTranspose(3, (3,3), activation='sigmoid')  # Ausgabe [0,1]
])

# ======================
# 4️ Autoencoder zusammensetzen
# ======================
input_img = tf.keras.Input(shape=(128, 128, 3))
encoded = encoder(input_img)
decoded = decoder(encoded)
autoencoder = tf.keras.Model(input_img, decoded)

#autoencoder.compile(optimizer='adam', loss='mse',  metrics=['r2_score'])
autoencoder.compile(optimizer='adam', loss='mse')

# ======================
# 5️ Training
# ======================
history = autoencoder.fit(X, X, epochs=10, batch_size=32, verbose=1)

#getrennt save und load
#encoder.save("encoder_model.h5")
#decoder.save("decoder_model.h5")

#encoder = load_model("encoder_model.h5")
#decoder = load_model("decoder_model.h5")

# Variante B – den gesamten Autoencoder speichern
# autoencoder.save("autoencoder_model.h5")
# autoencoder = load_model("autoencoder_model.h5")

# Zugriff auf Encoder und Decoder
# encoded_imgs = autoencoder.layers[1](X)  # Beispielhafter Zugriff

# Kontrolltest – sind Encoder/Decoder wirklich trainiert?
#before = encoder.get_weights()[0].copy()
#autoencoder.fit(X, X, epochs=1, batch_size=32)
#after = encoder.get_weights()[0]
#print("Weights changed:", not np.allclose(before, after))



# ======================
# 6️  Komprimierung: Features speichern
# ======================
compressed_features = encoder.predict(X)
print("Komprimierte Repräsentationen:", compressed_features.shape)

# Beispielsweise speichern:
np.save("compressed_features.npy", compressed_features)
print("Features gespeichert in compressed_features.npy")

#  die Bilder laden:
compressed_features = np.load("compressed_features.npy")
print("Geladene Feature-Shape:", compressed_features.shape)

# ======================
# 7️. Rekonstruktion über Decoder
# ======================
reconstructed = decoder.predict(compressed_features)
print("Rekonstruierte Bilder:", reconstructed.shape)

# ======================
# 8️. Beispielanzeige
# ======================
n = 3
plt.figure(figsize=(10, 4))
for i in range(n):
    # Original
    ax = plt.subplot(2, n, i + 1)
    plt.imshow(X[i])
    plt.title("Original")
    plt.axis("off")
    
    # Rekonstruktion
    ax = plt.subplot(2, n, i + 1 + n)
    plt.imshow(reconstructed[i])
    plt.title("Rekonstruiert")
    plt.axis("off")

plt.tight_layout()
plt.show()

```
