# warum sind LogisticRegression() und DecisionTreeClassifier() auch geeignet für iris daten?
## 1. LogisticRegression

Obwohl der Name "Regression" drin steckt, ist es ein Klassifikator (für diskrete Klassen).

Er modelliert die Wahrscheinlichkeit einer Klasse mit der logistischen Funktion.

Für mehr als zwei Klassen (wie beim Iris-Datensatz: 3 Klassen) verwendet Scikit-Learn die multinomiale logistische Regression (One-vs-Rest oder Softmax).

Da die Iris-Daten schön trennbar sind, kann Logistic Regression die Klassen sehr gut trennen.

Vorteil: liefert direkt Wahrscheinlichkeiten (predict_proba), die interpretierbar sind.

## 2. DecisionTreeClassifier

Ein Entscheidungsbaum teilt den Merkmalsraum rekursiv in Regionen ein (z. B. „Sepal length > 5.2?“).

Sehr gut geeignet, wenn nichtlineare Entscheidungsgrenzen gebraucht werden.

Beim Iris-Datensatz sind die Klassen teilweise auch nichtlinear getrennt, was der Baum gut abbilden kann.

Vorteil: Entscheidungsbäume sind leicht interpretierbar („Wenn-Blätter-Regeln“).

## 3. Vergleich zu k-NN

k-NN: sehr intuitiv, funktioniert gut, aber kann bei großen Datenmengen langsam werden.

LogisticRegression: lineares Modell, effizient, gute Basislinie für viele Probleme.

DecisionTreeClassifier: nichtlinear, interpretiert leicht, aber kann bei kleinen Daten leicht überfitten.

## 4. Warum beim Iris-Datensatz geeignet?

Iris ist ein kleiner, „sauberer“ Datensatz mit klar getrennten Klassen.

Viele Klassifikatoren (k-NN, Logistic Regression, Decision Trees, SVMs, Random Forests …) erreichen hier fast perfekte Genauigkeit.

Deshalb wird Iris oft als Benchmark-Datensatz für Klassifikationsalgorithmen verwendet.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier, RadiusNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

# Iris-Daten laden
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Liste von Modellen
models = [
    KNeighborsClassifier(n_neighbors=3),
    RadiusNeighborsClassifier(radius=1.0),
    LogisticRegression(max_iter=200),
    DecisionTreeClassifier()
]

# Modelle trainieren und auswerten
for model in models:
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    print(f"Modell: {model.__class__.__name__:<25} | Accuracy: {acc:.3f}")
```
