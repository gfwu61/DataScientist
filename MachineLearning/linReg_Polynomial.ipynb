
# PolynomialFeatures-Instanz


```python
import numpy as np
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()

from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression

# mehr dimension [x1,x2], interaction for x1,x2  
np.random.seed(42)
n_rows=20
X = np.random.randint(0, 10, size=(n_rows,2)) # 2 dimension matrix
print(X)
# 
poly = PolynomialFeatures(2, interaction_only=False, include_bias=False) # get the  instance: x1 + x2 + x1^2 +x1*x2+ x2^2: max. potenz ^2
X= poly.fit_transform(X)  # return  X, expand the 2 columns to 4 clumns and 1 bias colun
print("X", X)
a= np.array([1.5, -2., 1., 2.0, 1.2]) # Koefizient
y= X @ a + 1.2*np.random.rand(n_rows) # add noise
print("y", y)
lin_model = LinearRegression(fit_intercept=True)
lin_model.fit(X,y)

print("R^2 Score:", lin_model.score(X, y))

y1= lin_model.predict(X)
y2= y1-y
print("y3", np.sum(y2*y2)/ n_rows )
```

```python
print("Model slope:    ", lin_model.coef_)
print("Model intercept:", lin_model.intercept_)
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(1)

x = 10 * rng.random(50)        # statt rand()
y = np.sin(x) + 0.1 * rng.normal(size=50)  # statt randn()

plt.scatter(x, y)
plt.show()

```

```python

print("x.shape,y.shape", x.shape,y.shape)
#x = 10 * rng.rand(50)
# PolynomialFeatures(7)
X1= x[:, None]
cols= 12
for cols in range(2,cols):
    X=X1.copy()
    #poly = PolynomialFeatures(7, interaction_only=False)
    poly_model= PolynomialFeatures(cols, interaction_only=False, include_bias=False)
    X= poly_model.fit_transform(X)
    print(X.shape)

# train the line
    model = LinearRegression(fit_intercept=True)
    model.fit(X, y)
    r2_scoore= model.score(X, y)
    print(r2_scoore)


```

```python
print("R^2 Score:", model.score(X, y))
print("Model slope:    ", model.coef_)
print("Model intercept:", model.intercept_)
```
