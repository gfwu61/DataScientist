# WordNet in Python
Die wichtigsten Merkmale:

    -> Synsets (Synonym Sets): Das Herzstück von WordNet. Wörter werden nicht einzeln gespeichert, sondern in Gruppen von Synonymen zusammengefasst, die dasselbe Konzept beschreiben (z. B. {Auto, PKW, Kraftwagen}).
    -> Semantische Beziehungen: WordNet verknüpft diese Konzepte durch logische Beziehungen:
        Hypernyme: Oberbegriffe (z. B. "Möbel" ist Oberbegriff von "Stuhl").
        Hyponyme: Unterbegriffe (z. B. "Dackel" ist Unterbegriff von "Hund").
        Meronyme: Teil-Ganzes-Beziehungen (z. B. "Rad" ist Teil eines "Fahrrads").
    -> Einsatzzwecke: Es wird verwendet, um Wortbedeutungen zu unterscheiden (Wortbedeutungs-Disambiguierung), Ähnlichkeiten zwischen Begriffen zu berechnen oder als Basis für Sentiment-Analysen (wie bei SentiWordNet). 

    -> GermaNet (Universität Tübingen): Das bekannteste deutsche Pendant, das ähnlich wie WordNet aufgebaut ist.
    
    

```python
import nltk
from nltk.corpus import wordnet as wn

# Einmaliger Download der Datenbank
nltk.download('wordnet')
nltk.download('omw-1.4') # Open Multilingual Wordnet für Deutsch
```

```python
# Synsets für "dog" finden
syns = wn.synsets("dog")
# Die Kette nach oben verfolgen:

# 1. Das erste Synset für "dog" holen
dog = wn.synsets("dog")[0]
print(f"Startpunkt: {dog.name()} ({dog.definition()})")
print("-" * 30)

# 2. Den ersten Oberbegriff (Hypernym) finden
hyper = dog.hypernyms()[0]
print(f"1. Oberbegriff: {hyper.name()} (Haustier)")

# 3. Den nächsten Oberbegriff finden
next_hyper = hyper.hypernyms()[0]
print(f"2. Oberbegriff: {next_hyper.name()} (Hundeartige)")

# 4. Den übernächsten Oberbegriff finden
next_next_hyper = next_hyper.hypernyms()[0]
print(f"3. Oberbegriff: {next_next_hyper.name()} (Raubtier)")


print(f"Bedeutung: {syns[0].definition()}")
print(f"Beispiel: {syns[0].examples()}")

# Hypernyme (Oberbegriffe) finden
print(f"Oberbegriff: {syns[0].hypernyms()[0].name()}")
```

```python
from nltk.corpus import wordnet as wn

# Alle Synsets für das Wort "dog" abrufen
synsets = wn.synsets("dog")

print(f"Gefundene Bedeutungen für 'dog': {len(synsets)}\n")

for syn in synsets:
    # Der Name des Konzepts und seine Definition
    print(f"1. Konzept: {syn.name()} ({syn.definition()})")
    
    # Alle Synonyme (Lemmas) in diesem Set ausgeben
    synonyms = [lemma.name() for lemma in syn.lemmas()]
    print(f"2.  -> Synonyme: {synonyms}")
    print("-" * 30)
```
