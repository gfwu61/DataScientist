```python
import string

# Liste der Satzzeichen: !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
punctuation = set(string.punctuation)
print("punctuation\n",punctuation)
# Kombinierter Filter: Weder Stopword noch Satzzeichen
tokens = ["this", "Movie", "is", "very", "good","."]

filtered = [w for w in tokens if w not in stop_words and w not in punctuation]
print(filtered)
tokens = ["this", "Movie", "is", "very", "good",".#"]
filtered = [w for w in tokens if w not in stop_words and w not in punctuation]
print(filtered)

tokens = ["this", "Movie", "is", "very", "good",".#"]
# .lower() stellt sicher, dass auch "This" (groß) als Stopword "this" (klein) erkannt wird
filtered = [w.lower() for w in tokens if w.lower() not in stop_words and w not in punctuation]
print(filtered)


tokens = ["this", "Movie", "is", "very", "good",".#"]
# Filtert Wörter raus, die NUR aus Sonderzeichen bestehen (wie ".#", "!!!", "...")
filtered = [w for w in tokens if w not in stop_words and not all(char in punctuation for char in w)]
print(filtered)
```

# Lemmatization (besser als Stemming)

Reduktion auf lexikalische Grundform (Lemma) mithilfe von Wortarten.

    better → good
    running → run
    mice → mouse

    Für Sentiment Analysis & NLP-Tasks fast immer Lemmatization bevorzugen.

```python
from nltk.stem import WordNetLemmatizer
#import nltk
# nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize("running", pos="v"))
print(lemmatizer.lemmatize("better", pos="a"))
```

# Beispiel-Datensatz
Wir verwenden einen Mini-Datensatz (positiv / negativ):

```python
texts = [
    "I love this movie. It is amazing!",
    "This film was terrible and boring.",
    "Absolutely fantastic experience.",
    "I hate this product. Very bad quality.",
    "Best purchase ever!",
    "Worst thing I have bought."
]

labels = [1, 0, 1, 0, 1, 0]  # 1 = positiv, 0 = negativ
```

# Preprocessing mit NLTK
Wir machen:

Tokenization

Lowercasing

Stopword-Handling (vorsichtig!)

Lemmatization

```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

#nltk.download('punkt')
#nltk.download('stopwords')
#nltk.download('wordnet')

stop_words = set(stopwords.words("english"))

# WICHTIG: Negationen behalten!
stop_words.discard("not")

lemmatizer = WordNetLemmatizer()

def preprocess(text):
    tokens = word_tokenize(text.lower())
    print("tokens:", tokens)
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
        if word not in stop_words
        and word not in string.punctuation
    ]

    print(tokens)
    return " ".join(tokens)

processed_texts = [preprocess(t) for t in texts]

print(processed_texts)
```

```python
from nltk.tokenize import PunktSentenceTokenizer, sent_tokenize

# Ein Text mit einer speziellen Abkürzung, die NLTK normal nicht kennt
train_text = "Fr.Dr. Müller kommt heute. Fr.Dr. Meier kommt morgen."
test_text = "Guten Tag, Fr.Dr. Schmidt ist da."

# 1. Der Standard-Weg (macht Fehler)
print("Standard-Ergebnis:")
print(sent_tokenize(test_text))
# Ergebnis: ['Guten Tag, Fr.Dr.', 'Schmidt ist da.'] <- Falsch getrennt!

# 2. Der PunktSentenceTokenizer (lernt aus dem train_text)
custom_tokenizer = PunktSentenceTokenizer(train_text)

# Jetzt wenden wir den "gelernten" Tokenizer an
print("\nCustom-Ergebnis (nach Training):")
print(custom_tokenizer.tokenize(test_text))
# Ergebnis: ['Guten Tag, Fr.Dr. Schmidt ist da.'] <- Richtig zusammengehalten!
```

```python
from nltk.tokenize import PunktSentenceTokenizer

# 1. Ein großer "Lern-Text" (Training Data)
# Hier zeigen wir dem Tokenizer viele Beispiele, wie Abkürzungen genutzt werden.
training_data = """
Prof.Dr. Schmidt arbeitet an der Uni. 
Ca. 500 Studenten hören zu bzw. schreiben mit.
Prof.Dr. Meier ist ebenfalls dort. 
Es kostet ca. 10 Euro Eintritt bzw. ist für Studenten frei.
St.Pauli ist ein Stadtteil. Die U.S.A. sind groß.
"""

# 2. Den Tokenizer mit diesen Daten trainieren
custom_tokenizer = PunktSentenceTokenizer(training_data)

# 3. Ein neuer, unbekannter Text (Test Data)
# Hier prüfen wir, ob er gelernt hat, dass "Prof.Dr." und "ca." kein Satzende sind.
new_text = "Guten Morgen! Prof.Dr. Weber kommt ca. um 10 Uhr zu uns."

# 4. Anwenden
sentences = custom_tokenizer.tokenize(new_text)

for i, s in enumerate(sentences, 1):
    print(f"Satz {i}: {s}")
```

# Lösung: Manuell Abkürzungen hinzufügen (ohne den Punkt am Ende!)

```python
from nltk.tokenize import PunktSentenceTokenizer

# 1. Den Standard-Tokenizer laden
tokenizer = PunktSentenceTokenizer()

# 2. Manuell Abkürzungen hinzufügen (ohne den Punkt am Ende!)
extra_abbreviations = ['prof.dr', 'ca', 'bzw', 'st']
for abbr in extra_abbreviations:
    tokenizer._params.abbrev_types.add(abbr)

# 3. Testen
test_text = "Guten Morgen! Prof.Dr. Weber kommt ca. um 10 Uhr zu uns."
sentences = tokenizer.tokenize(test_text)

for i, s in enumerate(sentences, 1):
    print(f"Satz {i}: {s}")
```

# TF-IDF Vektorisierung

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(ngram_range=(1,2))  # Uni + Bigram
X = vectorizer.fit_transform(processed_texts)

print("Feature-Anzahl:", len(vectorizer.get_feature_names_out()))
print("TF-IDF Matrix Shape:", X.shape)
```

# Klassifikation mit PyTorch (Logistic Regression)


```python
import torch
import torch.nn as nn
import torch.optim as optim

X_tensor = torch.tensor(X.toarray(), dtype=torch.float32)
y_tensor = torch.tensor(labels, dtype=torch.float32).unsqueeze(1)
```

# Modell definieren

```python
class LogisticRegression(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.linear = nn.Linear(input_dim, 1)
    
    def forward(self, x):
        return torch.sigmoid(self.linear(x))
print("X_tensor.shape:", X_tensor.shape)
model = LogisticRegression(X_tensor.shape[1])
```

# Training

```python
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

for epoch in range(200):
    optimizer.zero_grad()
    
    outputs = model(X_tensor)
    loss = criterion(outputs, y_tensor)
    
    loss.backward()
    optimizer.step()
    
    if epoch % 50 == 0:
        print(f"Epoch {epoch}, Loss: {loss.item():.4f}")
```

```python
# Evaluation
```

```python
with torch.no_grad():
    predictions = model(X_tensor)
    predicted_classes = (predictions > 0.5).float()
    print("predicted_classes:", predicted_classes)
    
accuracy = (predicted_classes == y_tensor).float().mean()

print("Accuracy:", accuracy.item())
```
