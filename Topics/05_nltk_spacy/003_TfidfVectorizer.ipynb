# Vektorisierung mit TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

texts = [
    "I love this movie",
    "I hate this movie",
    "This movie is great",
    "This movie is terrible"
]

labels = [1, 0, 1, 0]  # 1 = positiv, 0 = negativ

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)
print("X", X)
print(vectorizer.get_feature_names_out())

import pandas as pd

# 1. Die Namen der Spalten (Wörter) abrufen
feature_names = vectorizer.get_feature_names_out()

# 2. Die Matrix in ein "dichtes" Format bringen und in einen DataFrame umwandeln
df = pd.DataFrame(X.toarray(), columns=feature_names, index=[f"Satz {i}" for i in range(len(texts))])

print(df)




clf = LogisticRegression()
clf.fit(X, labels)

coef_df = pd.DataFrame(clf.coef_, columns=vectorizer.get_feature_names_out())
print("coef_df:\n", coef_df)


test = vectorizer.transform(["I love this terrible movie"])
print("test text", test)
print("prediction resiult", clf.predict(test))
```
