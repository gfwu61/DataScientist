```python
from transformers import pipeline

# Lädt ein spezialisiertes deutsches Sentiment-Modell
# Es erkennt: 'positive', 'negative' oder 'neutral'
model_name = "oliverguhr/german-sentiment-bert"
sentiment_pipeline = pipeline("sentiment-analysis", model=model_name)

ntv_news = """
Die Nachricht klingt dramatisch, und sie ist es auch: 
In einem nur aus zwei Sätzen bestehenden Statement teilte der staatliche Energiekonzern Qatar Energy mit, 
dass er nach "militärischen Angriffen" auf zwei seiner Standorte die Produktion von LNG "eingestellt" habe.
"""
ntv_news1="""
Die Nachricht klingt dramatisch, und sie ist es auch: 
In einem nur aus zwei Sätzen bestehenden Statement teilte der staatliche Energiekonzern Qatar Energy mit, 
dass er nach "militärischen Angriffen" auf zwei seiner Standorte die Produktion von LNG "eingestellt" habe. Katar lieferte bis zu diesem Montag rund ein Fünftel des weltweiten Flüssiggasbedarfs. 
Zum ersten Mal in drei Jahrzehnten stellt Qatar Energy seinen Betrieb nun ein.

Die Reaktion an den globalen Finanzmärkten folgte innerhalb von Minuten: Die Gas-Futures - Terminkontrakte für Gaslieferungen in den kommenden Monaten - in Europa sprangen in die Höhe.
Angesichts der Blockade der für Exporte aus der Golfregion unverzichtbaren Straße von Hormus durch den Krieg stiegen auch die Ölpreise weiter.
Warnungen vor einem starken Anstieg der Inflation und einem Einbruch der Konjunktur folgten. 
"""


result = sentiment_pipeline(ntv_news)
print(f"Ergebnis: {result[0]['label']} (Score: {result[0]['score']:.4f})")

result = sentiment_pipeline(ntv_news1)
print(f"Ergebnis: {result[0]['label']} (Score: {result[0]['score']:.4f})")

test_text = "Ich bin absolut begeistert von der neuen Technologie, das ist fantastisch!"
result = sentiment_pipeline(test_text)
print(f"Ergebnis: {result[0]['label']} (Score: {result[0]['score']:.4f})")
print(type(result[0]))

test_text = "Der Film war okay, aber nicht überragend."
result = sentiment_pipeline(test_text)
print(f"Ergebnis: {result[0]['label']} (Score: {result[0]['score']:.4f})")
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# VADER initialisieren
analyzer = SentimentIntensityAnalyzer()

# Drei Varianten des gleichen Gefühls
texts = [
    "The movie is good.",                # 1. Neutral/Normal
    "The movie is GOOD!",                # 2. Großschreibung + Ausrufezeichen
    "The movie is extremely GOOD!!!"     # 3. Verstärker + mehr Ausrufezeichen
]

for t in texts:
    score = analyzer.polarity_scores(t)
    print(f"Text: '{t}'")
    print(f"  -> Compound Score: {score['compound']:.4f}")
```

# GoogleTranslator

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
from deep_translator import GoogleTranslator
import nltk

# 1. Übersetzer konfigurieren (Deutsch -> Englisch)
translator = GoogleTranslator(source='de', target='en')

# 2. VADER initialisieren
# Falls Fehler: nltk.download('vader_lexicon')
analyzer = SentimentIntensityAnalyzer()

# 3. Test-Funktion
def check_sentiment(text_de):
    # Übersetzen
    text_en = translator.translate(text_de)
    #print("text_en:", text_en)
    # Analysieren
    scores = analyzer.polarity_scores(text_en)
    print("scores:", scores)
    print()
    
    status = "Positiv" if scores['compound'] >= 0.05 else "Negativ" if scores['compound'] <= -0.05 else "Neutral"
    
    print(f"Original (DE): {text_de}")
    print(f"Übersetzt (EN): {text_en}")
    print(f"Ergebnis: {status} (Score: {scores['compound']})")
    print("-" * 30)

# Ausführen
check_sentiment("Die Bosch-Schulung war absolut hervorragend!")
check_sentiment("Der Prozess ist leider viel zu kompliziert und langsam.")
```
