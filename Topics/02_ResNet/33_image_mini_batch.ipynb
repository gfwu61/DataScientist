# CNN

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

# 1. Setup: 1000 Bilder (Dummy-Daten), Batch Size 32
images = torch.randn(1000, 3, 64, 64) # [N, C, H, W]
print("images:", images[0][0][0][0:10])
labels = torch.randint(0, 10, (1000,)) # 10 Klassen; [min=0, max=9(<10), 1000-Werte von 0 bis 9]
print("labels.shape:", labels.shape)
print("labels:", labels[0:10])
dataset = TensorDataset(images, labels)
loader = DataLoader(dataset, batch_size=32, shuffle=True)

model = nn.Sequential(nn.Flatten(), nn.Linear(3*64*64, 10))
optimizer = optim.SGD(model.parameters(), lr=0.01)
criterion = nn.CrossEntropyLoss()

# 2. Training Loop
num_epochs = 5 # Wir gehen 5-mal durch alle 1000 Bilder

for epoch in range(num_epochs):
    print(f"--- Epoch {epoch+1} beginnt ---")
    
    # Der Loader liefert uns insgesamt 32 Batches pro Epoch
    for batch_idx, (data, target) in enumerate(loader):
        
        # WICHTIG: Gradienten vom vorherigen Batch löschen
        optimizer.zero_grad()
        
        # Forward Pass
        output = model(data)
        loss = criterion(output, target)
        
        # Backward Pass (berechnet Gradienten)
        loss.backward()
        
        # Gewichte anpassen
        optimizer.step()
        
        if batch_idx % 10 == 0:
            print(f"Batch {batch_idx}: Verarbeite Bilder {batch_idx*32} bis {(batch_idx+1)*32}")
            print("output.shap :", output.shape)
            #print(" target:", target)
            print("output[0:2] :", output[0:2])

print("\nTraining abgeschlossen!")
```
