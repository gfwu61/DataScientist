#  Mittelwert und Standardabweichung für einen Ordner voll eigener Bilder automatisch berechnen

Wichtige Details:

ImageFolder: 

Diese Klasse von torchvision erwartet, dass deine Bilder in Unterordnern liegen (z.B. ./dein_ordner/klasse_a/bild1.jpg).

Resize: Da eigene Bilder oft unterschiedliche Größen haben, ist transforms.Resize wichtig, damit sie im DataLoader zu Batches zusammengefügt werden können.



```python
import torch
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

# 1. Nur ToTensor nutzen (um Werte von 0 bis 1 zu erhalten)
# Ersetze './dein_ordner' durch den Pfad zu deinen Bildern
transform = transforms.Compose([
    transforms.Resize((64, 64)), 
    transforms.ToTensor()
])
dataset = datasets.ImageFolder(root='./dein_ordner', transform=transform)
loader = DataLoader(dataset, batch_size=64, shuffle=False)

def get_mean_and_std(loader):
    mean = 0.0
    std = 0.0
    total_images = 0

    for images, _ in loader:
        # images shape: [batch_size, channels, height, width]
        batch_samples = images.size(0) 
        images = images.view(batch_samples, images.size(1), -1) # Flatten H und W
        mean += images.mean(2).sum(0)
        std += images.std(2).sum(0)
        total_images += batch_samples

    mean /= total_images
    std /= total_images

    return mean, std

mean, std = get_mean_and_std(loader)

print(f"Berechneter Mittelwert: {mean}")
print(f"Berechnete Std-Abweichung: {std}")
```
