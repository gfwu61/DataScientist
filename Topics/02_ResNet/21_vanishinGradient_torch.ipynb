# PyTorch Mini-Experiment (echte Zahlen!):  Vanishing Gradient
Tiefer linearer Stack mit kleinen Gewichten=0.5


```python
import torch
import torch.nn as nn

torch.manual_seed(0)

# tiefes lineares Netz
layers = []
depth = 10
for _ in range(depth):
    layer = nn.Linear(1, 1, bias=False)
    layer.weight.data.fill_(0.5)  # kleine Gewichte=0.0, alle gleich
    layers.append(layer)

model = nn.Sequential(*layers)

# Input
x = torch.tensor([[1.0]], requires_grad=True)

# Forward
y = model(x)

# Backward
y.backward()

print("Output:", y.item())
print("Gradient wrt x:", x.grad.item())
```

```python
# Nach y.backward()
print("x.grad:", x.grad.item())
print("Erster Layer grad:", layers[0].weight.grad.item())
print("Erster Layer grad:", layers[1].weight.grad.item())
print("Erster Layer grad:", layers[2].weight.grad.item())
print("Letzter Layer grad:", layers[-1].weight.grad.item())

# Check: Sollte ~0.0009765625 sein
assert abs(x.grad.item() - 0.5**10) < 1e-10
```
