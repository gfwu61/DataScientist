# ResNet50

```python
import numpy as np
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image

# 1. Modell mit vortrainierten ImageNet-Gewichten laden
model = ResNet50(weights='imagenet')

# 2. Bild laden und auf die Zielgröße von 224x224 anpassen
img_path = 'dein_testbild.jpg' # Pfad zu deinem Bild
img = image.load_img(img_path, target_size=(224, 224))

# 3. Vorverarbeitung: Bild in Array umwandeln und Batch-Dimension hinzufügen
x = image.img_to_array(img)
x = np.expand_dims(x, axis=0) # Macht aus (224, 224, 3) -> (1, 224, 224, 3)
x = preprocess_input(x)       # Normalisierung speziell für ResNet50

# 4. Vorhersage treffen
preds = model.predict(x)

# 5. Ergebnis dekodieren (Top-3 Vorhersagen anzeigen)
print('Vorhersage:', decode_predictions(preds, top=3)[0])
```
