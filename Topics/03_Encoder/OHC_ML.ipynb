```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import pandas as pd

# Beispiel-Daten
df = pd.DataFrame({
    'alter': [25, 45, 35, 20],
    'stadt': ['Berlin', 'München', 'Berlin', 'Hamburg'],
    'gekauft': [0, 1, 0, 1]
})

X = df.drop('gekauft', axis=1)
y = df['gekauft']

# 1. Definition der Spalten-Transformationen
# handle_unknown='ignore' ist entscheidend für neue Kategorien in Testdaten
preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), ['alter']),
    ('cat', OneHotEncoder(handle_unknown='ignore'), ['stadt'])
])

# 2. Erstellung der Pipeline (Vorverarbeitung + Modell)
clf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression())
])

# 3. Train-Test-Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# 4. Training: Die Pipeline lernt Kategorien NUR aus den Trainingsdaten
clf_pipeline.fit(X_train, y_train)

# 5. Vorhersage: Nutzt die gelernten Kategorien für die Testdaten
predictions = clf_pipeline.predict(X_test)
```
