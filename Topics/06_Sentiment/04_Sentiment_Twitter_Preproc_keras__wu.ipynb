```python
# -*- coding: utf-8 -*-
"""
Created on Fri Dec 10 09:32:17 2021

@author: Bernd Ebenhoch
"""

# Datensatz von:
# https://www.kaggle.com/abhi8923shriv/sentiment-analysis-dataset
# Veröffentlicht von: Abhishek Shrivastava
# License: CC0: Public Domain


import tensorflow as tf
from sklearn.isotonic import IsotonicRegression
import matplotlib.pyplot as plt
from sklearn.calibration import calibration_curve
from sklearn.metrics import recall_score
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from tensorflow import keras
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split


print('\n'*3+'-'*15+'Data loading and preprocessing'+'-'*15+'\n'*3)

# Daten laden
# Die Datei beinhaltet ein paar Sonderzeichen, diese können mit
# encoding='utf-8' geladen werden
Data = pd.read_csv('sentiment_Analyse.csv',
                   delimiter=',', encoding='utf-8')


# Anzahl der Datenpunkte ausgeben
print('Total number of datapoints:', len(Data))

# Das gesamte DataFrame in NumPy konvertieren
Data = Data.to_numpy()

# Die Sentimentbeurteilung in Spalte 0 als Label festlegen
# Die Texte der Tweets in Spalte 5 als Merkmal festlegen
y = Data[:, 0].astype('float32')/4  # Label=0 oder 1
X = Data[:, 5]

# Texte bereinigen
# Es werden alle Benutzernamen die mit @ beginnen entfernt
# Zusätzlich werden Sonderzeichen im XML-Format entfernt
for i in range(len(X)):
    sentence = " ".join(filter(lambda x: x[0] != '@', X[i].split()))
    sentence = sentence.replace('&ampPOOOOOLL', ' ')  # &
    sentence = sentence.replace('&quotPOOOOOLL', ' ')  # "
    sentence = sentence.replace('&ltPOOOOOLL', ' ')  # <
    sentence = sentence.replace('&gtPOOOOOLL', ' ')  # >
    sentence = sentence.replace('POOOOOLL', ' ')

    X[i] = sentence

print(X[0:5])

# Datensatz in Trainings- und Validationdaten aufteilen
X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.2, random_state=0)


# Ein Tokenizer zerlegt die Sätze in einzelne Wörter und weist jedem Wort durch ein
# Dictionary eine Zahl (Integer) zu
# Die Zahlen werden nach Worthäufigkeit vergeben, das häufigste Wort erhält die
# niedrigste Zahl
# Mit num_words kann das Dictionary auf die häufigsten Wörter reduziert werden
# Filters entfernt die angegebenen Sonderzeichen
sequence_length = 25
num_words = 10000
tokenizer = keras.layers.TextVectorization(
    max_tokens=num_words,
    standardize='lower_and_strip_punctuation',
    split='whitespace',
    ngrams=None,
    output_mode='int',
    output_sequence_length=sequence_length,
    pad_to_max_tokens=True,
    vocabulary=None,
    idf_weights=None,
    sparse=False,
    ragged=False,
    encoding='utf-8',
    name=None,
)

# Den Tokenizer auf die Trainingsdaten fitten
tokenizer.adapt(X_train)

# 10 000, wie num_words wegen pad_to_max_tokens=True
vocabulary = tokenizer.get_vocabulary()
vocab_size = len(tokenizer.get_vocabulary())

print( "vocabulary len:", len(vocabulary) , vocab_size ) 

# tokenizer_model erzeugen
tokenizer_model = keras.models.Sequential()
tokenizer_model.add(tokenizer)

# X_train: Text in ID(Token-ID) umwandeln
X_train = tokenizer_model.predict(X_train, verbose=False, batch_size=40000)
X_val = tokenizer_model.predict(X_val, verbose=False, batch_size=40000)


print('Shapes von X_train und X_val:', X_train.shape, X_val.shape)
print('X_train und X_val:\n', X_train[0:1], X_val[0:1])

# test new_data = np.array(['Job por'], dtype='object')
# auffüllen mit 0 wenn nicht genuge Wörter vorhanden
new_data = np.array(['Job por'], dtype='object')
new_data = tokenizer_model.predict(new_data)
print(new_data)



# Trainingsmodel erzeugen
# Modell anlegen
# sequence_length = 25
# vocab_size=10000

print(X_train.shape[1:])

model = keras.models.Sequential()
# In der Embedding-Schicht entsteht pro Wort ein 3-dimensionaler Vektor
#model.add(keras.layers.Input((25,)))
model.add(keras.layers.Input( X_train.shape[1:] ))# immer korrekt
model.add(keras.layers.Embedding(vocab_size, 3, mask_zero=True))
#model.add(keras.layers.LSTM(128, return_sequences=True))  # Output: (batch, 25, 128)
#model.add(keras.layers.LSTM(64, return_sequences=False))  # Output: (batch, 64)

model.add(keras.layers.LSTM(64, return_sequences=True))  # Output: (batch, 25, 128)
model.add(keras.layers.LSTM(32, return_sequences=False))  # Output: (batch, 64)


# model.add(keras.layers.Flatten())
# model.add(keras.layers.Reshape((max_len*3,)))  # ist das selbe wie Flatten
# model.add(keras.layers.LSTM(100, return_sequences=False))
model.add(keras.layers.Dense(1, activation='sigmoid'))

model.summary()


model.compile(optimizer=keras.optimizers.Adam(learning_rate=0.01),
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.fit(X_train, y_train , epochs=10, batch_size=500, 
            validation_data=(X_val, y_val), verbose=True) # epoch=1, 500


# print()
# print(model.layers[0].weights[0].numpy())

loss, accuracy = model.evaluate(X_train, y_train, verbose=0)
# print()
print('Accuracy: %f' % (accuracy*100))

# Word Vektoren grafisch darstellen

print( len(model.layers))
layer = model.layers[0]  # Embedding

print("Name:", layer.name)
#print("Input Shape:", layer.input_shape)
#print("Output Shape:", layer.output_shape)
print("Trainable:", layer.trainable)
print("Weights Shape:", [w.shape for w in layer.weights])
print("Config:", layer.get_config())

# vocab_size=10 000
X_embed = model.layers[0](np.arange(vocab_size))

print(X_embed[0:10])

# 1. Dimension der Wort-Vektoren auf X-Achse,
# 2. Dimension auf y-Achse, 3. auf die Z-Achse abbilden
ax = plt.figure().add_subplot(projection='3d')

print_size= 30
ax.view_init(10, 20)
ax.scatter3D(X_embed[:print_size, 0], X_embed[:print_size, 1], X_embed[:print_size, 2], color='red')

for i, vec in enumerate(X_embed[:print_size]):
    ax.text(vec[0], vec[1], vec[2], vocabulary[i])


plt.show()

# =================
new_data = np.array(['bad','sad', 'good', 'satisfied', 'happay'], dtype='object')

new_data = tokenizer_model.predict(new_data)

# Aussage durch das Model nach dem Training
print(" value of the words")
print(model.predict(new_data))

#======================================
# zusammenfassung von tokenizer und model: das gleiche machen wie der code oben
application_model = keras.models.Sequential()
application_model.add(keras.layers.Input((1,), dtype=tf.string))
application_model.add(tokenizer)
application_model.add(model)

application_model.summary()

#application_model.save('application_model.keras')
#application_model = keras.models.load_model('application_model.keras')

example = np.array(['great', 'satisfied', 'sad', ' I am not happy because it is not shining','I love Trump'], dtype='object')
result = application_model.predict(example)

print()
print(example)
print()
print('Result:')
print(result)

```
