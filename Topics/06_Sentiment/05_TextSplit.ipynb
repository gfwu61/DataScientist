```python
import tensorflow as tf
from tensorflow.keras.layers import TextVectorization

texts = ["apple orange banana lemon", 
         "grape lime mango"]
vectorize_layer = TextVectorization(
    max_tokens=20,
    output_mode='multi_hot',
    pad_to_max_tokens=True
)
vectorize_layer.adapt(texts)
print(vectorize_layer.get_vocabulary())

output = vectorize_layer(texts)
print(output)
print(output.shape)
```

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np

# Beispielhafte Eingabesequenz mit Token-IDs (Index-Werte zwischen 0 und 9)
# Batchgröße 2, max_len=4 Tokens pro Beispiel
input_sequences = np.array([
  [1, 2, 3, 4],
  [4, 3, 2, 0]  # 0 als Padding
])

vocab_size = 10
max_len = 4

model = keras.Sequential()
model.add(keras.layers.Input(shape=(max_len,)))
model.add(keras.layers.Embedding(input_dim=vocab_size, output_dim=3, mask_zero=True))
model.add(keras.layers.Flatten())
#model.add(keras.layers.GlobalAveragePooling1D())  # unterstützt Masken
model.add(keras.layers.Dense(1, activation='sigmoid'))



# Modell zusammenfassen (als Übersicht)
model.summary()

# Beispiel Forward Pass: Ausgabe des Embeddings für die Eingabesequenzen
output = model(input_sequences)
print("Model output:")
print(output.numpy())
```

```python
import numpy as np
import pandas as pd

# ---- 1️⃣ Beispiel-Daten ----
# Stell dir vor: 10 Tage Daten mit 6 Spalten
# Spalten: [sales, temp, rain, humidity, public_holiday, weekday]
data = np.arange(10 * 6)
print(data)
data=data.reshape(10,6)
print("Ursprüngliche Daten (erste 10 Zeilen):\n", data)

# Forecast: wir wollen 3 Tage in die Zukunft vorhersagen
forecast_length = 2

# ---- 2️⃣ Feiertags- und Wochentagsspalten verschieben ----
# Wir verschieben ALLE Spalten ab Index 4 um 3 Zeilen nach oben
data[:-forecast_length, 4:] = data[forecast_length:, 4:]

print("\nNach Verschiebung (ab Spalte 4 um 3 Zeilen):\n", data)
```
