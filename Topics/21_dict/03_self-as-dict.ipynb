# 1. Klasse verhält sich wie ein Dictionary
## Schlüsselidee: __getitem__

- ➡️ self ist eine Klasseninstanz,
-  aber sie delegiert den Zugriff an ein internes Dictionary.

```python
class User:
    def __init__(self, name, age):
        self._data = {
            "name": name,
            "age": age
        }

    def __getitem__(self, key):
        print(f"getitem called with key={key}")
        return self._data[key]

    def func(self):
        return self["name"]

u = User("Du", 42)
print(u.func())
```

# 2.  Klasse erbt von dict: dict is a class constuctor for dictionary
## Sehr verbreitet z. B. in Frameworks (FastAPI, Pydantic, Django intern).
## The dict() Constructor: dict(name = "John", age = 36, country = "Norway")

➡️ self IST ein Dictionary,

➡️ aber gleichzeitig eine Klasseninstanz.

User  ──▶  dict  ──▶  object
- u = User(...)
- isinstance(u, User)   # True
- isinstance(u, dict)   # True
- isinstance(u, object) # True
self === u

und u ist:

    eine User-Instanz

    ein Dictionary

    ein Mapping

    ein Objekt

Alles gleichzeitig.

```python
class User(dict):
    def func(self):
        return self["name"]
# diffrence of dict-definition:{"name":"Du", "age":35}
# and function call: name="Du", age=35
mydict= {"name":"Du", "age":35}
#u = User(name="Du", age=35) or
u = User(mydict)
print(u.func())
```

```python
thisdict = dict(name = "John", age = 36, country = "Norway")
print(thisdict)
```

```python
class DebugDict:
    def __init__(self):
        self.store = {"name": "Du"}

    def __getitem__(self, key):
        print(f"getitem called with key={key}")
        return self.store[key]

    def func(self):
        return self["name"]

d = DebugDict()
print(d.func())
```
