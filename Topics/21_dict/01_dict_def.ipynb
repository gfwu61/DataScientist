### dict( line=, form=); comma "," as seperator in (key1=val1, key2=val2,key3=val3)
(line=val1, ...), key1-> line, val1 -> a dict obejct

```python
marker= dict(
            line= dict(
                color='black',  # Linienfarbe
                width=1         # Liniendicke
            ),
            form="Red"
         )
marker    
```
