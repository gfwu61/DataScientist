# kern_size=1, stride=1

```python
import torch
import torch.nn as nn

# Einfacher Input: 1x3x4x4 (1 Bild, 3 Kanäle, 4x4 Pixel)
x = torch.arange(48.0).reshape(1, 3, 4, 4)
print("Input (3 Kanäle, 4x4):")
#print(x[:,:,0:2,0:2])  # Erste 2x2 Pixel zeigen
print(x )

conv = nn.Conv2d(3, 3, kernel_size=1)
print("\nGewichte (uninitialisiert):")
#print(conv.weight[0, :, 0, 0])  # Erstes Gewicht zeigen
print(conv.weight)  # Erstes Gewicht zeigen

out = conv(x)
print(f"\nOutput shape: {out.shape}")  # [1, 3, 4, 4] ✓
print("Output erste 2x2 Pixel:")
print(out[:,:,0:2,0:2])
```

# kern_size=1, stride=2 (for downsampling)

```python
import torch
import torch.nn as nn

# Einfacher Input: 1x3x4x4 (1 Bild, 3 Kanäle, 6x6 Pixel)
x = torch.arange(108.0).reshape(1, 3, 6, 6)
print("Input (3 Kanäle, 6x6):")
print(x)  # Erste 2x2 Pixel zeigen

conv = nn.Conv2d(3, 3, kernel_size=1,stride=2)
print("\nGewichte (uninitialisiert):")
#print(conv.weight[0, :, 0, 0])  # Erstes Gewicht zeigen
print(conv.weight)  # Erstes Gewicht zeigen

out = conv(x)
print(f"\nOutput shape: {out.shape}")  # [1, 3, 4, 4] ✓

print(out)
```

# AdaptiveAvgPool2d
Input H=7, Output h=2 → Kernel_size = ceil(7/2) = 4, Stride = floor(7/2)=3

```python
import torch
import torch.nn as nn

# ResNet layer4 Output
x = torch.arange(49.0).reshape(1, 1, 7, 7)  # 1 Kanal, 7×7
print("Input 7×7:")
print(x.squeeze())

pool = nn.AdaptiveAvgPool2d((2, 2))
out = pool(x)
print(f"\nOutput shape: {out.shape}")
print("Output 2×2:")
print(out.squeeze())
```

```python
x = torch.randn(1, 12, 7, 7)      # layer4 Output
out = nn.AdaptiveAvgPool2d((1, 1))(x)              # → [1, 12, 1, 1]
print(out.shape)                   # Bleibt immer 1×1 egal ob 7×7 oder 100×100!
print(out)
```
