# einvollständigen, lauffähigen PyTorch-Code für ein CNN mit Conv2d, genau nach deinen Vorgaben:

Input X = (40, 3, 128, 128)

4 Conv-Layer

Pooling-Layer nach jeder Conv

out_channels = 2 (Klassifikation)

Epochs = 5

Mini-Batch = 4

inkl. Training + echter Konsolen-Ausgabe

Input (3×128×128)

>  ↓ Conv(3→16) + ReLU + MaxPool
> 
>  ↓ Conv(16→32) + ReLU + MaxPool
> 
>  ↓ Conv(32→64) + ReLU + MaxPool
> 
>  ↓ Conv(64→128) + ReLU + MaxPool
> 
>  ↓ Flatten
> 
>  ↓ Linear → 2 Klassen

128 → 64 → 32 → 16 → 8


```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
```

```python
# -----------------------------
# Daten
# -----------------------------
X = torch.randn(40, 3, 128, 128)     # 40 Bilder
y = torch.randint(0, 2, (40,))       # 2 Klassen (0/1)

dataset = TensorDataset(X, y)
loader = DataLoader(dataset, batch_size=4, shuffle=True)
```

## CNN mit 4 Conv + Pooling

```python
class CNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),          # 128 → 64

            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),          # 64 → 32

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),          # 32 → 16

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)           # 16 → 8
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 8 * 8, 2)   # out_c = 2
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x
```

## Modell, Loss, Optimizer

```python
model = CNN()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3)
```

## Training (5 Epochen, Mini-Batch = 4)

```python
# -----------------------------
# Training
# -----------------------------
epochs = 30

for epoch in range(epochs):
    total_loss = 0.0

    for batch_x, batch_y in loader:
        optimizer.zero_grad()

        logits = model(batch_x)
        loss = criterion(logits, batch_y)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch [{epoch+1}/{epochs}] - Loss: {total_loss:.4f}")
```
