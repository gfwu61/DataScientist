```python
import psutil

# Anzahl der echten physischen Kerne
physical_cores = psutil.cpu_count(logical=False)

# Anzahl der logischen Threads (inkl. Hyper-Threading)
logical_threads = psutil.cpu_count(logical=True)

print(f"Physische Kerne: {physical_cores}")
print(f"Logische Threads: {logical_threads}")
```

```python
import os
import multiprocessing

# Gibt die Anzahl der logischen Prozessoren (Threads) zurück
threads = os.cpu_count()
print(f"Dein System hat {threads} Threads.")

# Alternative über multiprocessing
print(f"Threads via multiprocessing: {multiprocessing.cpu_count()}")
```

```python
import torch

# Begrenze PyTorch auf z.B. 8 Kerne (statt 20+)
torch.set_num_threads(8)
torch.set_num_interop_threads(8)

print(f"Threads begrenzt auf: {torch.get_num_threads()}")
```
